# Question 1
##1. Partition Function

For a quantum harmonic oscillator, the energy eigenvalues are

$$
E_j=\hbar\omega\left(j+\frac12\right), \qquad j=0,1,2,\ldots
$$

The canonical partition function is defined as

$$
Z(T)=\sum_{j=0}^{\infty}e^{-\frac{E_j}{k_BT}}.
$$

Substituting the energy eigenvalues,

$$
Z(T)=\sum_{j=0}^{\infty}
\exp\left[-\frac{\hbar\omega\left(j+\frac12\right)}{k_BT}\right].
$$

Using the exponential property

$$
e^{-(a+b)}=e^{-a}e^{-b},
$$

the partition function becomes

$$
Z(T)
=
e^{-\frac{\hbar\omega}{2k_BT}}
\sum_{j=0}^{\infty}
e^{-j\frac{\hbar\omega}{k_BT}}.
$$

Let

$$
r=e^{-\frac{\hbar\omega}{k_BT}}.
$$

Since

$$
\hbar>0,\qquad
\omega>0,\qquad
T>0,
$$

we have

$$
0<r<1.
$$

Therefore, the infinite summation is a convergent geometric series.

From the geometric series identity

$$
\sum_{j=0}^{\infty}r^j
=
\frac{1}{1-r},
\qquad |r|<1,
$$

we obtain

$$
\sum_{j=0}^{\infty}
e^{-j\frac{\hbar\omega}{k_BT}}
=
\frac{1}
{1-e^{-\frac{\hbar\omega}{k_BT}}}.
$$

Hence,

$$
Z(T)
=
e^{-\frac{\hbar\omega}{2k_BT}}
\left(
\frac{1}
{1-e^{-\frac{\hbar\omega}{k_BT}}}
\right).
$$

Therefore, the canonical partition function of the quantum harmonic oscillator is

$$
{
Z(T)
=
\frac{
e^{-\frac{\hbar\omega}{2k_BT}}
}
{
1-e^{-\frac{\hbar\omega}{k_BT}}
}
}
$$

This agrees with the partition function obtained using the canonical ensemble discussed in the lecture notes. The partition function completely characterizes the thermodynamic properties of the quantum harmonic oscillator and will be used to derive the Helmholtz free energy, internal energy, entropy, and heat capacity in the following sections.




## 2. Thermodynamic Potentials

The Helmholtz free energy is defined as

$$
F(T)=-k_BT\ln Z.
$$

From Question 1, the partition function is

$$
Z(T)=
\frac{
e^{-\frac{\hbar\omega}{2k_BT}}
}
{
1-e^{-\frac{\hbar\omega}{k_BT}}
}.
$$

Taking the natural logarithm,

$$
\ln Z
=
\ln\left(
e^{-\frac{\hbar\omega}{2k_BT}}
\right)
-
\ln\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right).
$$

Using the logarithm identity

$$
\ln(e^x)=x,
$$

gives

$$
\ln Z
=
-\frac{\hbar\omega}{2k_BT}
-
\ln\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right).
$$

Substituting this expression into the Helmholtz free energy,

$$
F(T)
=
-k_BT
\left[
-\frac{\hbar\omega}{2k_BT}
-
\ln\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right)
\right].
$$

Multiplying through by $-k_BT$,

$$
F(T)
=
\frac{\hbar\omega}{2}
+
k_BT
\ln\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right).
$$

Therefore, the Helmholtz free energy is

$$
F(T)
=
\frac{\hbar\omega}{2}
+
k_BT
\ln\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right).
$$

### Internal Energy

According to the canonical ensemble discussed in the lecture notes, the internal energy is

$$
U(T)
=
-\frac{\partial\ln Z}{\partial\beta},
$$

where

$$
\beta=\frac{1}{k_BT}.
$$

Expressing the partition function in terms of $\beta$,

$$
Z
=
\frac{
e^{-\beta\hbar\omega/2}
}
{
1-e^{-\beta\hbar\omega}
}.
$$

Taking the natural logarithm,

$$
\ln Z
=
-\frac{\beta\hbar\omega}{2}
-
\ln\left(
1-e^{-\beta\hbar\omega}
\right).
$$

Differentiating with respect to $\beta$,

$$
\frac{\partial\ln Z}{\partial\beta}
=
-\frac{\hbar\omega}{2}
-
\frac{\partial}{\partial\beta}
\ln\left(
1-e^{-\beta\hbar\omega}
\right).
$$

Using the chain rule,

$$
\frac{\partial}{\partial\beta}
\ln\left(
1-e^{-\beta\hbar\omega}
\right)
=
\frac{1}{1-e^{-\beta\hbar\omega}}
\left(
\hbar\omega e^{-\beta\hbar\omega}
\right).
$$

Therefore,

$$
\frac{\partial\ln Z}{\partial\beta}
=
-\frac{\hbar\omega}{2}
-
\frac{
\hbar\omega e^{-\beta\hbar\omega}
}
{
1-e^{-\beta\hbar\omega}
}.
$$

Using the definition of internal energy,

$$
U(T)
=
-\frac{\partial\ln Z}{\partial\beta},
$$

we obtain

$$
U(T)
=
\frac{\hbar\omega}{2}
+
\frac{
\hbar\omega e^{-\beta\hbar\omega}
}
{
1-e^{-\beta\hbar\omega}
}.
$$

Since

$$
\frac{e^{-x}}{1-e^{-x}}
=
\frac{1}{e^x-1},
$$

the internal energy simplifies to

$$
U(T)
=
\frac{\hbar\omega}{2}
+
\frac{\hbar\omega}
{e^{\beta\hbar\omega}-1}.
$$

Finally, substituting

$$
\beta=\frac{1}{k_BT},
$$

gives

$$
U(T)
=
\frac{\hbar\omega}{2}
+
\frac{\hbar\omega}
{
e^{\frac{\hbar\omega}{k_BT}}-1
}.
$$

The first term,

$$
\frac{\hbar\omega}{2},
$$

represents the zero-point energy of the quantum harmonic oscillator, while the second term represents the thermal energy due to excitation of higher quantum states.

## 3. Entropy

From Question 2, the Helmholtz free energy is

$$
F(T)
=
\frac{\hbar\omega}{2}
+
k_BT
\ln\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right).
$$

According to the thermodynamic relation introduced in the lecture notes, the entropy is

$$
S(T)
=
-
\left(
\frac{\partial F}{\partial T}
\right).
$$

Differentiating the first term,

$$
\frac{\partial}{\partial T}
\left(
\frac{\hbar\omega}{2}
\right)
=
0.
$$

Therefore,

$$
\frac{\partial F}{\partial T}
=
\frac{\partial}{\partial T}
\left[
k_BT
\ln
\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right)
\right].
$$

Using the product rule,

$$
\frac{\partial F}{\partial T}
=
k_B
\ln
\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right)
+
k_BT
\frac{\partial}{\partial T}
\ln
\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right).
$$

Now evaluate the second derivative.

Using the chain rule,

$$
\frac{\partial}{\partial T}
\ln
\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right)
=
\frac{1}
{1-e^{-\frac{\hbar\omega}{k_BT}}}
\cdot
\frac{\partial}
{\partial T}
\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right).
$$

Since

$$
\frac{\partial}
{\partial T}
e^{-\frac{\hbar\omega}{k_BT}}
=
e^{-\frac{\hbar\omega}{k_BT}}
\frac{\hbar\omega}{k_BT^2},
$$

we obtain

$$
\frac{\partial}{\partial T}
\ln
\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right)
=
-
\frac{
\hbar\omega
e^{-\frac{\hbar\omega}{k_BT}}
}
{
k_BT^2
\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right)
}.
$$

Substituting this result,

$$
\frac{\partial F}{\partial T}
=
k_B
\ln
\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right)
-
\frac{
\hbar\omega
e^{-\frac{\hbar\omega}{k_BT}}
}
{
T
\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right)
}.
$$

Therefore,

$$
S(T)
=
-
\frac{\partial F}{\partial T}.
$$

Hence,

$$
S(T)
=
-
k_B
\ln
\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right)
+
\frac{
\hbar\omega
e^{-\frac{\hbar\omega}{k_BT}}
}
{
T
\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right)
}.
$$

Using the identity

$$
\frac{e^{-x}}
{1-e^{-x}}
=
\frac{1}
{e^x-1},
$$

the entropy can be written as

$$
S(T)
=
-
k_B
\ln
\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right)
+
\frac{\hbar\omega}
{
T
\left(
e^{\frac{\hbar\omega}{k_BT}}-1
\right)
}.
$$

### Verification of the Thermodynamic Identity

The thermodynamic identity is

$$
U=F+TS.
$$

Substituting the expressions for the Helmholtz free energy and entropy,

$$
F+TS
=
\left[
\frac{\hbar\omega}{2}
+
k_BT
\ln
\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right)
\right]
+
T
\left[
-
k_B
\ln
\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right)
+
\frac{\hbar\omega}
{
T
\left(
e^{\frac{\hbar\omega}{k_BT}}-1
\right)
}
\right].
$$

Expanding the expression,

$$
F+TS
=
\frac{\hbar\omega}{2}
+
k_BT
\ln
\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right)
-
k_BT
\ln
\left(
1-e^{-\frac{\hbar\omega}{k_BT}}
\right)
+
\frac{\hbar\omega}
{
e^{\frac{\hbar\omega}{k_BT}}-1
}.
$$

The logarithmic terms cancel,

$$
F+TS
=
\frac{\hbar\omega}{2}
+
\frac{\hbar\omega}
{
e^{\frac{\hbar\omega}{k_BT}}-1
}.
$$

From Question 2,

$$
U(T)
=
\frac{\hbar\omega}{2}
+
\frac{\hbar\omega}
{
e^{\frac{\hbar\omega}{k_BT}}-1
}.
$$

Therefore,

$$
F+TS=U.
$$

Hence, the derived expressions satisfy the thermodynamic identity,

$$
U=F+TS,
$$

which verifies the consistency of the canonical ensemble formulation presented in the lecture notes.

##4. Heat Capacity and Hyperbolic Form

The constant-volume heat capacity is defined as

$$
C_V
=
\left(
\frac{\partial U}{\partial T}
\right)_V.
$$

From Question 2, the internal energy of the quantum harmonic oscillator is

$$
U(T)
=
\frac{\hbar\omega}{2}
+
\frac{\hbar\omega}
{
e^{\frac{\hbar\omega}{k_BT}}-1
}.
$$

Since the first term is a constant,

$$
\frac{\partial}{\partial T}
\left(
\frac{\hbar\omega}{2}
\right)
=
0.
$$

Therefore,

$$
C_V
=
\frac{\partial}{\partial T}
\left[
\frac{\hbar\omega}
{
e^{\frac{\hbar\omega}{k_BT}}-1
}
\right].
$$

Let

$$
x
=
\frac{\hbar\omega}{k_BT}.
$$

Then,

$$
U
=
\frac{\hbar\omega}{2}
+
\frac{\hbar\omega}{e^x-1}.
$$

Using the chain rule,

$$
C_V
=
\frac{\partial U}{\partial x}
\cdot
\frac{\partial x}{\partial T}.
$$

First,

$$
\frac{\partial x}{\partial T}
=
-
\frac{\hbar\omega}{k_BT^2}.
$$

Next,

$$
\frac{\partial}{\partial x}
\left(
\frac{1}{e^x-1}
\right)
=
-
\frac{e^x}{(e^x-1)^2}.
$$

Hence,

$$
\frac{\partial U}{\partial x}
=
-
\hbar\omega
\frac{e^x}{(e^x-1)^2}.
$$

Substituting into the chain rule,

$$
C_V
=
\left(
-
\hbar\omega
\frac{e^x}{(e^x-1)^2}
\right)
\left(
-
\frac{\hbar\omega}{k_BT^2}
\right).
$$

Therefore,

$$
C_V
=
\frac{(\hbar\omega)^2}{k_BT^2}
\frac{e^x}{(e^x-1)^2}.
$$

Since

$$
x=\frac{\hbar\omega}{k_BT},
$$

the heat capacity becomes

$$
C_V
=
k_B
\left(
\frac{\hbar\omega}{k_BT}
\right)^2
\frac{
e^{\frac{\hbar\omega}{k_BT}}
}
{
\left(
e^{\frac{\hbar\omega}{k_BT}}-1
\right)^2
}.
$$

### Hyperbolic Form

Using the hyperbolic identity

$$
\sinh(y)
=
\frac{e^y-e^{-y}}{2},
$$

let

$$
y
=
\frac{x}{2}.
$$

Then,

$$
\sinh\left(\frac{x}{2}\right)
=
\frac{
e^{x/2}-e^{-x/2}
}{2}.
$$

Squaring both sides,

$$
\sinh^2\left(\frac{x}{2}\right)
=
\frac{
\left(
e^{x/2}-e^{-x/2}
\right)^2
}{4}.
$$

Expanding,

$$
\left(
e^{x/2}-e^{-x/2}
\right)^2
=
e^x-2+e^{-x}.
$$

Hence,

$$
4\sinh^2\left(\frac{x}{2}\right)
=
e^x-2+e^{-x}.
$$

Multiplying by $e^x$,

$$
4e^x
\sinh^2\left(\frac{x}{2}\right)
=
e^{2x}-2e^x+1.
$$

Recognizing that

$$
e^{2x}-2e^x+1
=
(e^x-1)^2,
$$

gives

$$
(e^x-1)^2
=
4e^x
\sinh^2\left(\frac{x}{2}\right).
$$

Therefore,

$$
\frac{e^x}{(e^x-1)^2}
=
\frac{1}
{
4\sinh^2\left(\frac{x}{2}\right)
}.
$$

Substituting this identity into the expression for the heat capacity,

$$
C_V
=
k_B
x^2
\left(
\frac{1}
{
4\sinh^2\left(\frac{x}{2}\right)
}
\right).
$$

Since

$$
x
=
\frac{\hbar\omega}{k_BT},
$$

we have

$$
\frac{x^2}{4}
=
\left(
\frac{\hbar\omega}{2k_BT}
\right)^2.
$$

Therefore,

$$
C_V
=
k_B
\left(
\frac{\hbar\omega}{2k_BT}
\right)^2
\frac{1}
{
\sinh^2
\left(
\frac{\hbar\omega}{2k_BT}
\right)
}.
$$

This is the required compact hyperbolic expression for the constant-volume heat capacity of a quantum harmonic oscillator. It is equivalent to the exponential form but is often preferred because of its compactness and symmetry.

##5. High-Temperature Limit of the Heat Capacity

From Question 4, the constant-volume heat capacity of the quantum harmonic oscillator is

$$
C_V
=
k_B
\left(
\frac{\hbar\omega}{2k_BT}
\right)^2
\frac{1}
{
\sinh^2
\left(
\frac{\hbar\omega}{2k_BT}
\right)
}.
$$

To determine the high-temperature limit, let

$$
x
=
\frac{\hbar\omega}{2k_BT}.
$$

As the temperature increases,

$$
T \rightarrow \infty,
$$

we have

$$
x \rightarrow 0.
$$

For very small values of \(x\), the Taylor series expansion of the hyperbolic sine function is

$$
\sinh(x)
=
x
+
\frac{x^3}{6}
+
\frac{x^5}{120}
+\cdots.
$$

Since \(x\) is very small, the higher-order terms become negligible. Therefore,

$$
\sinh(x)
\approx
x.
$$

Hence,

$$
\sinh^2(x)
\approx
x^2.
$$

Substituting this approximation into the expression for the heat capacity,

$$
C_V
=
k_B
x^2
\frac{1}{\sinh^2(x)}
\approx
k_B
x^2
\frac{1}{x^2}.
$$

The factor \(x^2\) cancels, giving

$$
C_V
\approx
k_B.
$$

Therefore,

$$
\lim_{T\rightarrow\infty}C_V
=
k_B.
$$

### Physical Interpretation

At very high temperatures,

$$
k_BT
\gg
\hbar\omega,
$$

which means the thermal energy is much larger than the spacing between adjacent quantum energy levels. As a result, many vibrational energy states become accessible, and the oscillator behaves like a classical harmonic oscillator.

According to the classical equipartition theorem discussed in the lecture notes, each quadratic degree of freedom contributes

$$
\frac{1}{2}k_BT
$$

to the average energy.

A one-dimensional harmonic oscillator has two quadratic energy terms:

- Kinetic energy
- Potential energy

Therefore, the average internal energy becomes

$$
U
=
\frac{1}{2}k_BT
+
\frac{1}{2}k_BT
=
k_BT.
$$

The heat capacity is then

$$
C_V
=
\frac{dU}{dT}
=
\frac{d(k_BT)}{dT}
=
k_B.
$$

This agrees exactly with the high-temperature limit obtained mathematically.

Hence, the quantum harmonic oscillator approaches the classical equipartition result as the temperature becomes very large, giving

$$
C_V
=
k_B.
$$

This result demonstrates the correspondence between quantum statistical mechanics and classical thermodynamics, where quantum effects become negligible at sufficiently high temperatures.

# Question 2 - The Maximum Entropy Distribution under Mean and Variance Constraints

Consider a continuous random variable

$$
X\in(-\infty,\infty).
$$

The probability density function is denoted by

$$
p(x).
$$

The Shannon-Jaynes entropy for a continuous random variable is

$$
H[p]
=
-\int_{-\infty}^{\infty}
p(x)
\ln p(x)\,dx.
$$

We wish to maximize the entropy subject to the following constraints.

### Constraint 1: Normalization

$$
\int_{-\infty}^{\infty}
p(x)\,dx=1.
$$

### Constraint 2: Mean

$$
\int_{-\infty}^{\infty}
xp(x)\,dx=\mu.
$$

### Constraint 3: Variance

$$
\int_{-\infty}^{\infty}
(x-\mu)^2p(x)\,dx=\sigma^2.
$$

---

## Step 1: Construct the Lagrangian

Introduce the Lagrange multipliers

$$
\lambda_0,\;
\lambda_1,\;
\lambda_2.
$$

The Lagrangian is

$$
\begin{aligned}
\mathcal L
=&
-\int
p(x)\ln p(x)\,dx
\\
&
-\lambda_0
\left(
\int p(x)\,dx-1
\right)
\\
&
-\lambda_1
\left(
\int xp(x)\,dx-\mu
\right)
\\
&
-\lambda_2
\left(
\int (x-\mu)^2p(x)\,dx-\sigma^2
\right).
\end{aligned}
$$

---

## Step 2: Differentiate with Respect to the Probability Density

For maximum entropy,

$$
\frac{\delta\mathcal L}{\delta p(x)}
=
0.
$$

Differentiating,

$$
-\left(1+\ln p(x)\right)
-\lambda_0
-\lambda_1x
-\lambda_2(x-\mu)^2
=
0.
$$

Hence,

$$
\ln p(x)
=
-1
-\lambda_0
-\lambda_1x
-\lambda_2(x-\mu)^2.
$$

Exponentiating both sides,

$$
p(x)
=
\exp
\left(
-1
-\lambda_0
-\lambda_1x
-\lambda_2(x-\mu)^2
\right).
$$

Rearranging,

$$
p(x)
=
Ae^{-\lambda_1x}
e^{-\lambda_2(x-\mu)^2},
$$

where

$$
A=e^{-1-\lambda_0}.
$$

---

## Step 3: Apply the Mean Constraint

Since the required distribution has mean

$$
\mu,
$$

the probability density must be symmetric about

$$
x=\mu.
$$

Therefore,

$$
\lambda_1=0.
$$

Hence,

$$
p(x)
=
Ae^{-\lambda_2(x-\mu)^2}.
$$

---

## Step 4: Determine the Normalization Constant

The normalization condition gives

$$
\int_{-\infty}^{\infty}
Ae^{-\lambda_2(x-\mu)^2}
dx
=
1.
$$

Using the Gaussian integral,

$$
\int_{-\infty}^{\infty}
e^{-a(x-b)^2}dx
=
\sqrt{\frac{\pi}{a}},
$$

we obtain

$$
A
\sqrt{\frac{\pi}{\lambda_2}}
=
1.
$$

Therefore,

$$
A
=
\sqrt{\frac{\lambda_2}{\pi}}.
$$

---

## Step 5: Apply the Variance Constraint

The variance of the Gaussian distribution is

$$
\sigma^2
=
\frac{1}{2\lambda_2}.
$$

Therefore,

$$
\lambda_2
=
\frac{1}{2\sigma^2}.
$$

Substituting into the normalization constant,

$$
A
=
\sqrt{
\frac{1}{2\pi\sigma^2}
}.
$$

Hence,

$$
A
=
\frac{1}{\sqrt{2\pi\sigma^2}}.
$$

---

## Step 6: Final Maximum Entropy Distribution

Substituting the values of

$$
A
\quad\text{and}\quad
\lambda_2,
$$

the probability density becomes

$$
p(x)
=
\frac{1}{\sqrt{2\pi\sigma^2}}
\exp
\left(
-\frac{(x-\mu)^2}{2\sigma^2}
\right).
$$

Therefore, the probability distribution that maximizes entropy under the constraints of fixed mean and variance is

$$
X
\sim
\frac{1}{\sqrt{2\pi\sigma^2}}
\exp
\left(
-\frac{(x-\mu)^2}{2\sigma^2}
\right).
$$

---

## Conclusion

The Maximum Entropy Principle states that among all continuous probability distributions having the same mean and variance, the Gaussian distribution possesses the largest entropy. Therefore, it is the least biased probability distribution consistent with the available information.

This result is one of the fundamental results of statistical mechanics and information theory and agrees with the derivation presented in the lecture notes.

# Question 3 - Multiclass Logistic Regression: Automatic Classification of Emails

An e-commerce platform classifies customer support emails into one of two categories:

- **Finance** $(Y=0)$
- **Technical** $(Y=1)$

Each incoming email is represented by the binary feature vector

$$
\mathbf{x}
=
(x_1,x_2,x_3)^T,
$$

where

- $x_1=1$ if financial keywords are detected.
- $x_2=1$ if account access keywords are detected.
- $x_3=1$ if bug or malfunction keywords are detected.

The observed feature expectation constraints obtained from historical data are

$$
E[x_1I(Y=1)]
=
0.05,
$$

$$
E[x_2I(Y=1)]
=
0.15,
$$

and

$$
E[x_3I(Y=1)]
=
0.16.
$$

The objective is to derive the Maximum Entropy (MaxEnt) classifier, which is mathematically equivalent to Multiclass Logistic Regression.

## Step 1 - Conditional Entropy

According to the Maximum Entropy Principle discussed in the lecture notes, the least biased classifier is obtained by maximizing the conditional Shannon entropy

$$
H(Y|X)
=
-
\sum_x
P(x)
\sum_y
P(y|x)
\ln P(y|x).
$$

The optimization is performed subject to two conditions.

### 1. Probability Normalization

For every feature vector,

$$
\sum_y
P(y|x)
=
1.
$$

### 2. Feature Expectation Constraints

The learned classifier must reproduce the observed statistics

$$
E[x_1I(Y=1)]
=
0.05,
$$

$$
E[x_2I(Y=1)]
=
0.15,
$$

$$
E[x_3I(Y=1)]
=
0.16.
$$

These constraints ensure that the model agrees with the empirical observations while remaining as unbiased as possible.

## Step 2 - Construct the Lagrangian

To maximize the conditional entropy while satisfying the observed feature expectations, we introduce one Lagrange multiplier for each constraint.

Let

$$
\lambda_1,\lambda_2,\lambda_3
$$

be the multipliers corresponding to the three feature expectation constraints.

The Lagrangian is

$$
\begin{aligned}
\mathcal{L}
=&
-
\sum_xP(x)

\end{aligned}
$$

The Lagrange multipliers ensure that the resulting probability distribution satisfies all observed constraints while maximizing entropy.

## Step 3 - Differentiate the Lagrangian

To obtain the optimal conditional probability distribution, differentiate the Lagrangian with respect to

$$

and set the derivative equal to zero,

## Step 4 - Determine the Partition Function

# Question 4 - Asset Pricing and Portfolio Optimization

The objective is to determine the least-biased probability density function

$$
p(S_T)
$$

using the Maximum Entropy Principle.

According to the lecture notes, the differential entropy of a continuous probability density function is

$$
H[p]
=
-\int_0^\infty
p(S_T)
\ln p(S_T)
\,dS_T.
$$

The Maximum Entropy Principle states that among all probability density functions satisfying the observed market information, the distribution with the largest entropy is selected because it introduces the least amount of additional assumptions.

## Step 1 - Market Constraints

The probability density function must satisfy the following constraints.

### 1. Normalization

Since

$$
p(S_T)
$$

is a probability density function,

$$
\int_0^\infty
p(S_T)
\,dS_T
=
1.
$$

### 2. Expected Future Price

The expected future asset price is constrained by

=

\int_0^\infty
S_T
p(S_T)
\,dS_T
=

=

Therefore,

$$
\int_0^\infty
S_T^2
p(S_T)
\,dS_T
=
10600.

### 4. Logarithmic Constraint

The skewness information supplied by the options market is

$$

=

Hence,

$$
\int_0^\infty
\ln(S_T)
p(S_T)
\,dS_T
=
4.60.

These constraints summarize all the information available from the financial market. The Maximum Entropy method determines the least-biased probability density that satisfies all of these conditions simultaneously.


## Step 2 - Construct the Lagrangian

To maximize the entropy while satisfying the market constraints, introduce four Lagrange multipliers

$$
\lambda_0,\;
\lambda_1,\;
\lambda_2,\;
\lambda_3.

$$
\begin{aligned}

=&

\right).
\end{aligned}
$$

The first term represents the entropy to be maximized, while the remaining terms enforce the observed market constraints. The Lagrange multipliers determine how strongly each constraint influences the final probability density function.

## Step 3 - Differentiate the Lagrangian

To obtain the probability density function that maximizes entropy, take the functional derivative of the Lagrangian with respect to

-\lambda_1S_T
