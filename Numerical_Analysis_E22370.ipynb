## 1. Bias in the Maximum Likelihood Covariance Estimator

Consider a sequence of independent and identically distributed random vectors

$$
X_1,X_2,\ldots,X_n \in \mathbb{R}^{m}
$$

with population mean

$$
\boldsymbol{\mu}=E[X_i]
$$

and population covariance matrix

$$
\boldsymbol{\Sigma}
=
E[(X_i-\boldsymbol{\mu})(X_i-\boldsymbol{\mu})^T].
$$

The empirical sample mean is

$$
\widehat{\boldsymbol{\mu}}
=
\frac{1}{n}\sum_{i=1}^{n}X_i.
$$

Under the multivariate normal assumption, the Maximum Likelihood Estimator (MLE) of the covariance matrix is

$$
\widehat{\boldsymbol{\Sigma}}_{MLE}
=
\frac{1}{n}
\sum_{i=1}^{n}
(X_i-\widehat{\boldsymbol{\mu}})
(X_i-\widehat{\boldsymbol{\mu}})^T.
$$

Although this estimator maximizes the likelihood function, it does not provide an unbiased estimate of the true covariance matrix.

The scatter matrix

$$
\sum_{i=1}^{n}
(X_i-\widehat{\boldsymbol{\mu}})
(X_i-\widehat{\boldsymbol{\mu}})^T
$$

contains only \(n-1\) effective degrees of freedom because one degree of freedom is consumed when estimating the sample mean.

Therefore,

$$
E
\left[
\sum_{i=1}^{n}
(X_i-\widehat{\boldsymbol{\mu}})
(X_i-\widehat{\boldsymbol{\mu}})^T
\right]
=
(n-1)\boldsymbol{\Sigma}.
$$

Substituting this result into the MLE covariance estimator gives

$$
E[\widehat{\boldsymbol{\Sigma}}_{MLE}]
=
E\left[
\frac{1}{n}
\sum_{i=1}^{n}
(X_i-\widehat{\boldsymbol{\mu}})
(X_i-\widehat{\boldsymbol{\mu}})^T
\right]
$$

$$
=
\frac{1}{n}(n-1)\boldsymbol{\Sigma}
$$

$$
=
\frac{n-1}{n}\boldsymbol{\Sigma}.
$$

Since

$$
\frac{n-1}{n}<1,
$$

the estimator systematically underestimates the true population covariance matrix.

Hence, the bias is

$$
Bias
=
E[\widehat{\boldsymbol{\Sigma}}_{MLE}]
-
\boldsymbol{\Sigma}
$$

$$
=
\frac{n-1}{n}\boldsymbol{\Sigma}
-
\boldsymbol{\Sigma}
$$

$$
=
-\frac{1}{n}\boldsymbol{\Sigma}.
$$

Therefore, the Maximum Likelihood covariance estimator is biased for finite sample sizes.


## Elimination of Bias Using Bessel's Correction

To remove this bias, the covariance estimator is scaled using \(n-1\) instead of \(n\).

The unbiased sample covariance matrix is defined as

$$
\mathbf{S}
=
\frac{1}{n-1}
\sum_{i=1}^{n}
(X_i-\widehat{\boldsymbol{\mu}})
(X_i-\widehat{\boldsymbol{\mu}})^T.
$$

Taking expectation,

$$
E[\mathbf{S}]
=
\frac{1}{n-1}
E
\left[
\sum_{i=1}^{n}
(X_i-\widehat{\boldsymbol{\mu}})
(X_i-\widehat{\boldsymbol{\mu}})^T
\right].
$$

Substituting

$$
E
\left[
\sum_{i=1}^{n}
(X_i-\widehat{\boldsymbol{\mu}})
(X_i-\widehat{\boldsymbol{\mu}})^T
\right]
=
(n-1)\boldsymbol{\Sigma},
$$

we obtain

$$
E[\mathbf{S}]
=
\frac{1}{n-1}(n-1)\boldsymbol{\Sigma}
=
\boldsymbol{\Sigma}.
$$

Therefore,

$$
E[\mathbf{S}]
=
\boldsymbol{\Sigma},
$$

which proves that the sample covariance matrix \(\mathbf{S}\) is an unbiased estimator of the population covariance matrix.

# 2. Type I Error and Type II Error in Structural Health Monitoring

Consider the hypothesis test

$$
H_0:
\text{The structure remains healthy and operates normally}
$$

$$
H_1:
\text{The structure has undergone structural degradation or damage}.
$$

### Type I Error

A Type I Error occurs when the null hypothesis is rejected even though it is true.

$$
\alpha
=
P(\text{Reject } H_0 \mid H_0 \text{ is true}).
$$

In a structural health monitoring system, this corresponds to a **false alarm**, where a healthy structure is incorrectly classified as damaged.

Possible consequences include:

- Unnecessary inspections.
- Increased maintenance costs.
- Unplanned operational interruptions.
- Reduced confidence in the monitoring system.


### Type II Error

A Type II Error occurs when the null hypothesis is not rejected even though it is false.

$$
\beta
=
P(\text{Fail to Reject } H_0 \mid H_1 \text{ is true}).
$$

In practice, this means that actual structural damage exists but remains undetected.

Possible consequences include:

- Continued operation of a damaged structure.
- Progressive deterioration.
- Reduced structural reliability.
- Increased risk of catastrophic failure.

---

## Consequence of an Ultra-Conservative Threshold $\alpha = 0.0001\$

Suppose the significance level is reduced from a common value such as

$$
\alpha = 0.05
$$

to

$$
\alpha = 0.0001.
$$

This shifts the decision threshold further into the extreme tail of the reference distribution.

### Impact on Statistical Power

Because the rejection region becomes smaller,

- Type I Error decreases significantly.
- False alarms become extremely rare.
- Type II Error increases.
- The probability of detecting real damage decreases.

The statistical power of the test is

$$
Power = 1-\beta.
$$

Since \(\beta\) increases, the statistical power decreases.

As a result, the monitoring system becomes highly conservative and less sensitive to early-stage structural changes.

---

## Geometric Interpretation Using Confidence Ellipsoids

In multivariate statistical monitoring, the healthy operating region is represented by a confidence ellipsoid centred around the baseline mean.

The confidence level is

$$
1-\alpha.
$$

For

$$
\alpha = 0.0001,
$$

the confidence level becomes

$$
99.99\%.
$$

Consequently, the confidence ellipsoid expands and occupies a larger volume in feature space.

This larger ellipsoid encloses more observations and makes it harder for new measurements to be classified as abnormal.

Therefore:

- False alarms decrease.
- Detection sensitivity decreases.
- Small structural shifts may remain hidden within the enlarged healthy region.

Hence, although an ultra-conservative significance level improves robustness against false alarms, it simultaneously increases the risk of missing genuine structural degradation.

# Part B: Theoretical Extension (Asymptotic Distributions & Slutsky's Theorem)

The multivariate Central Limit Theorem (CLT) establishes the asymptotic behaviour of the sample mean vector. Consider a sequence of i.i.d. random vectors

$$
X_1,X_2,\ldots,X_n \in \mathbb{R}^{m}
$$

with population mean

$$
\boldsymbol{\mu}
=
E[X_i]
$$

and covariance matrix

$$
\boldsymbol{\Sigma}
=
Var(X_i).
$$

The empirical sample mean is

$$
\widehat{\boldsymbol{\mu}}_n
=
\frac{1}{n}
\sum_{i=1}^{n}X_i.
$$

According to the Lindberg-Lévy Multivariate Central Limit Theorem,

$$
\sqrt{n}
\left(
\widehat{\boldsymbol{\mu}}_n
-
\boldsymbol{\mu}
\right)
\xrightarrow{d}
\mathcal{N}
(\mathbf{0},\boldsymbol{\Sigma}),
$$

where \(\xrightarrow{d}\) denotes convergence in distribution.

This result states that as the sample size becomes very large, the scaled estimation error of the sample mean approaches a multivariate Gaussian distribution centered at zero.

---

## Standardization of the Limiting Distribution

Multiplying both sides by

$$
\boldsymbol{\Sigma}^{-1/2}
$$

gives

$$
\boldsymbol{\Sigma}^{-1/2}
\sqrt{n}
\left(
\widehat{\boldsymbol{\mu}}_n
-
\boldsymbol{\mu}
\right)
\xrightarrow{d}
\mathcal{N}
(\mathbf{0},\mathbf{I}),
$$

where

$$
\mathbf{I}
$$

is the identity matrix.

This transformation removes the covariance structure and converts the limiting distribution into a standard multivariate Gaussian distribution.

However, the population covariance matrix

$$
\boldsymbol{\Sigma}
$$

is unknown in practice and cannot be directly used.

---

## Consistency of the Sample Covariance Matrix

The unbiased sample covariance matrix is

$$
\mathbf{S}
=
\frac{1}{n-1}
\sum_{i=1}^{n}
(X_i-\widehat{\boldsymbol{\mu}}_n)
(X_i-\widehat{\boldsymbol{\mu}}_n)^T.
$$

By the Weak Law of Large Numbers,

$$
\mathbf{S}
\xrightarrow{p}
\boldsymbol{\Sigma},
$$

where \(\xrightarrow{p}\) denotes convergence in probability.

Therefore,

$$
\mathbf{S}^{-1/2}
\xrightarrow{p}
\boldsymbol{\Sigma}^{-1/2}.
$$

This shows that the sample covariance matrix is a consistent estimator of the true covariance matrix.

---

## Application of Slutsky's Theorem

Slutsky's Theorem states that if

$$
Y_n
\xrightarrow{d}
Y
$$

and

$$
A_n
\xrightarrow{p}
A,
$$

then

$$
A_nY_n
\xrightarrow{d}
AY.
$$

In this problem,

$$
Y_n
=
\sqrt{n}
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})
$$

and

$$
A_n
=
\mathbf{S}^{-1/2}.
$$

Since

$$
\sqrt{n}
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})
\xrightarrow{d}
\mathcal{N}
(\mathbf{0},\boldsymbol{\Sigma})
$$

and

$$
\mathbf{S}^{-1/2}
\xrightarrow{p}
\boldsymbol{\Sigma}^{-1/2},
$$

Slutsky's Theorem gives

$$
\mathbf{S}^{-1/2}
\sqrt{n}
(\widehat{\boldsymbol{\mu}}_n-\boldsymbol{\mu})
\xrightarrow{d}
\mathcal{N}
(\mathbf{0},\mathbf{I}).
$$

Thus the unknown population covariance matrix can be replaced by its empirical estimate without changing the asymptotic distribution.

---

## Practical Operational Distribution

Rearranging the above result yields

$$
\widehat{\boldsymbol{\mu}}_n
\approx
\mathcal{N}
\left(
\boldsymbol{\mu},
\frac{1}{n}\mathbf{S}
\right).
$$

Therefore, for sufficiently large sample sizes,

$$
{
\widehat{\boldsymbol{\mu}}_n
\sim
\mathcal{N}
\left(
\boldsymbol{\mu},
\frac{1}{n}\mathbf{S}
\right)
}
$$

which provides a practical statistical model for the uncertainty of the empirical mean.

This result is extremely important in structural health monitoring and industrial diagnostics because the true covariance matrix is never observable. By substituting the consistent sample covariance matrix into the asymptotic distribution, engineers can construct confidence ellipsoids, anomaly detection boundaries, and statistical control limits directly from measured data while preserving the asymptotic validity guaranteed by the Central Limit Theorem and Slutsky's Theorem.

```python
import numpy as np
import pandas as pd
import scipy.stats as stats

np.random.seed(42)

n_samples = 5000
n_features = 3

base_data = np.random.multivariate_normal(
    mean=[0.5, -0.2, 1.1],
    cov=[[0.09, 0.02, 0.01],
         [0.02, 0.06, 0.03],
         [0.01, 0.03, 0.05]],
    size=n_samples
)

base_data[4000:, 0] += 0.015
base_data[4000:, 2] += 0.010

df_strain = pd.DataFrame(
    base_data,
    columns=['Strain_Ch1', 'Strain_Ch2', 'Strain_Ch3']
)
```

```python
def verify_first_moment_homogeneity(df: pd.DataFrame, g_chunks: int = 5) -> dict:

    # ---------------------------------------------------
    # 1. Split data into equal consecutive chunks
    # ---------------------------------------------------
    chunks = np.array_split(df.values, g_chunks)

    n_total = len(df)
    m = df.shape[1]

    # Global mean
    mu_global = np.mean(df.values, axis=0).reshape(-1, 1)

    # ---------------------------------------------------
    # 2. Compute Within-Chunk Matrix (W)
    # ---------------------------------------------------
    W = np.zeros((m, m))

    for chunk in chunks:
        mu_chunk = np.mean(chunk, axis=0).reshape(-1, 1)

        for x in chunk:
            x = x.reshape(-1, 1)
            W += (x - mu_chunk) @ (x - mu_chunk).T

    # ---------------------------------------------------
    # 3. Compute Between-Chunk Matrix (B)
    # ---------------------------------------------------
    B = np.zeros((m, m))

    for chunk in chunks:
        n_j = len(chunk)

        mu_chunk = np.mean(chunk, axis=0).reshape(-1, 1)

        diff = mu_chunk - mu_global

        B += n_j * (diff @ diff.T)

    # ---------------------------------------------------
    # 4. Wilks' Lambda
    # ---------------------------------------------------
    Lambda = np.linalg.det(W) / np.linalg.det(W + B)

    # ---------------------------------------------------
    # 5. Bartlett's Chi-Square Approximation
    # ---------------------------------------------------
    chi2_calc = -(n_total - 1 - (m + g_chunks) / 2) * np.log(Lambda)

    df_chi2 = m * (g_chunks - 1)

    p_value = 1 - stats.chi2.cdf(chi2_calc, df=df_chi2)

    # ---------------------------------------------------
    # 6. Decision Rule
    # ---------------------------------------------------
    alpha = 0.05

    if p_value < alpha:
        conclusion = (
            "Reject H0: Evidence of mean shift detected. "
            "The first moment is not homogeneous."
        )
    else:
        conclusion = (
            "Fail to Reject H0: No significant mean shift detected. "
            "The first moment remains homogeneous."
        )

    return {
        "Wilks_Lambda": Lambda,
        "ChiSquare": chi2_calc,
        "Degrees_of_Freedom": df_chi2,
        "p_value": p_value,
        "Conclusion": conclusion
    }
```

```python
results = verify_first_moment_homogeneity(df_strain, g_chunks=5)

for k, v in results.items():
    print(f"{k}: {v}")
```

## Numerical Verification Results

The dataset was partitioned into five consecutive blocks of equal size. Using the MANOVA framework, the Within-Chunk matrix (W) and Between-Chunk matrix (B) were computed to evaluate first-moment homogeneity.

Wilks' Lambda was calculated as

$$
\Lambda = \frac{|W|}{|W+B|}
$$

and transformed into Bartlett's Chi-Square statistic. The corresponding p-value was compared with the significance level \(\alpha = 0.05\).

If the p-value is less than 0.05, the null hypothesis is rejected and a significant mean shift is detected. Otherwise, the null hypothesis cannot be rejected, indicating that the baseline mean remains statistically stable across time.

## 1. Covariance Transformation Under PCA

Let

$$
\widetilde{\mathbf{X}}_i
=
\mathbf{X}_i-\boldsymbol{\mu}
\in \mathbb{R}^{m}
$$

denote a centered random feature vector with population covariance matrix

$$
\boldsymbol{\Sigma}
=
E[\widetilde{\mathbf{X}}_i\widetilde{\mathbf{X}}_i^T].
$$

Since \(\boldsymbol{\Sigma}\) is symmetric and positive semi-definite, it admits the spectral decomposition

$$
\boldsymbol{\Sigma}
=
\mathbf{P}\boldsymbol{\Lambda}\mathbf{P}^T
$$

where

$$
\mathbf{P}
=
[\mathbf{p}_1,\mathbf{p}_2,\ldots,\mathbf{p}_m]
$$

contains orthonormal eigenvectors and

$$
\boldsymbol{\Lambda}
=
\text{diag}(\lambda_1,\lambda_2,\ldots,\lambda_m)
$$

contains the corresponding eigenvalues.

The PCA coordinate transformation is defined by

$$
\mathbf{Z}_i
=
\mathbf{P}^T\widetilde{\mathbf{X}}_i.
$$

The covariance matrix of the transformed vector is

$$
E[\mathbf{Z}_i\mathbf{Z}_i^T]
=
E[(\mathbf{P}^T\widetilde{\mathbf{X}}_i)
(\mathbf{P}^T\widetilde{\mathbf{X}}_i)^T].
$$

Using the transpose identity,

$$
E[\mathbf{Z}_i\mathbf{Z}_i^T]
=
\mathbf{P}^T
E[\widetilde{\mathbf{X}}_i\widetilde{\mathbf{X}}_i^T]
\mathbf{P}.
$$

Substituting the population covariance matrix,

$$
E[\mathbf{Z}_i\mathbf{Z}_i^T]
=
\mathbf{P}^T
\boldsymbol{\Sigma}
\mathbf{P}.
$$

Using the spectral decomposition,

$$
E[\mathbf{Z}_i\mathbf{Z}_i^T]
=
\mathbf{P}^T
(\mathbf{P}\boldsymbol{\Lambda}\mathbf{P}^T)
\mathbf{P}.
$$

Since

$$
\mathbf{P}^T\mathbf{P}
=
\mathbf{I},
$$

we obtain

$$
E[\mathbf{Z}_i\mathbf{Z}_i^T]
=
\boldsymbol{\Lambda}.
$$

Therefore,

$$
Cov(\mathbf{Z}_i)
=
\boldsymbol{\Lambda}.
$$

Because \(\boldsymbol{\Lambda}\) is diagonal,

$$
Cov(Z_{ij},Z_{ik})=0
\quad (j\neq k).
$$

Hence, all principal components are uncorrelated.

### Interpretation

The PCA transformation rotates the coordinate system so that the covariance matrix becomes diagonal. Each principal component represents an independent variance direction, and all cross-covariances are eliminated.

---

## 2. Conservation of Total Variance

The total variance of the original system is given by

$$
\text{tr}(\boldsymbol{\Sigma}).
$$

Substituting the spectral decomposition,

$$
\text{tr}(\boldsymbol{\Sigma})
=
\text{tr}
(\mathbf{P}\boldsymbol{\Lambda}\mathbf{P}^T).
$$

Using the cyclic property of the trace,

$$
\text{tr}
(\mathbf{P}\boldsymbol{\Lambda}\mathbf{P}^T)
=
\text{tr}
(\boldsymbol{\Lambda}\mathbf{P}^T\mathbf{P}).
$$

Since

$$
\mathbf{P}^T\mathbf{P}
=
\mathbf{I},
$$

we obtain

$$
\text{tr}(\boldsymbol{\Sigma})
=
\text{tr}(\boldsymbol{\Lambda}).
$$

Therefore,

$$
\text{tr}(\boldsymbol{\Sigma})
=
\sum_{j=1}^{m}\lambda_j.
$$

This proves that PCA preserves the total system variance.


### Explained Variance Ratio

If the first \(k\) principal components are retained, the cumulative explained variance ratio is

$$
\Phi(k)
=
\frac{\sum_{j=1}^{k}\lambda_j}
{\sum_{j=1}^{m}\lambda_j}.
$$

This quantity measures the fraction of total information retained in the reduced-dimensional representation.

### Residual Unexplained Variance

The remaining variance discarded by truncation is

$$
\Psi(k)
=
1-\Phi(k).
$$

Equivalently,

$$
\Psi(k)
=
\frac{\sum_{j=k+1}^{m}\lambda_j}
{\sum_{j=1}^{m}\lambda_j}.
$$

This residual variance corresponds to the information contained in the discarded subspace.

---

## 3. Orthogonal Subspace Decomposition

Retaining only the first \(k\) principal components, the centered observation can be decomposed as

$$
\widetilde{\mathbf{X}}_i
=
\mathbf{P}_k\mathbf{z}_{i,k}
+
\mathbf{P}_{m-k}\mathbf{z}_{i,m-k}.
$$

Define

$$
\widehat{\mathbf{X}}_i
=
\mathbf{P}_k\mathbf{z}_{i,k}
$$

as the reconstructed vector and

$$
\mathbf{e}_i
=
\mathbf{P}_{m-k}\mathbf{z}_{i,m-k}
$$

as the residual vector.

Thus,

$$
\widetilde{\mathbf{X}}_i
=
\widehat{\mathbf{X}}_i+\mathbf{e}_i.
$$

Since the retained and discarded eigenvectors are orthogonal,

$$
\widehat{\mathbf{X}}_i^T\mathbf{e}_i
=
0.
$$

Applying the Pythagorean theorem,

$$
\|\widetilde{\mathbf{X}}_i\|^2
=
\|\widehat{\mathbf{X}}_i\|^2
+
\|\mathbf{e}_i\|^2.
$$

Therefore, the total energy of the observation is partitioned into explained and residual components.


## Diagnostic Statistics

### Hotelling's \(T^2\)

The Hotelling's statistic measures variation within the retained principal component space:

$$
T_i^2
=
\sum_{j=1}^{k}
\frac{z_{ij}^2}{\lambda_j}.
$$

A large value of \(T_i^2\) indicates abnormal behaviour within the dominant structural subspace.

### Squared Prediction Error (SPE)

The residual statistic measures variation outside the retained subspace:

$$
Q_i
=
\|\mathbf{e}_i\|^2.
$$

Large SPE values indicate that new behaviour has appeared in directions not captured by the principal subspace.

### Structural Interpretation

- **Environmental or operational variations** primarily affect the dominant principal components and are more visible through Hotelling's \(T^2\).
- **Local damage, sensor faults, or previously unseen structural changes** generate energy in the residual subspace and are more effectively detected using the SPE statistic.

Therefore, Hotelling's \(T^2\) monitors changes within the modeled structure, whereas SPE monitors deviations outside the learned structural model.

```python
import numpy as np
import pandas as pd

class PCAOptimizationEngine:

    def __init__(self, df):
        self.df = df

    def compute_pca_metrics(self):

        # --------------------------------------------------
        # 1. Center data
        # --------------------------------------------------
        X = self.df.values

        mu_hat = np.mean(X, axis=0)

        X_centered = X - mu_hat

        n, m = X_centered.shape

        # --------------------------------------------------
        # 2. Unbiased covariance matrix
        # --------------------------------------------------
        S = np.cov(X_centered, rowvar=False, ddof=1)

        # --------------------------------------------------
        # 3. Eigen decomposition
        # --------------------------------------------------
        eigvals, eigvecs = np.linalg.eigh(S)

        idx = np.argsort(eigvals)[::-1]

        eigvals = eigvals[idx]
        eigvecs = eigvecs[:, idx]

        # --------------------------------------------------
        # 4. Principal component scores
        # --------------------------------------------------
        Z = X_centered @ eigvecs

        results = {}

        # --------------------------------------------------
        # 5. Evaluate k = 1,2,3
        # --------------------------------------------------
        for k in [1, 2, 3]:

            # Hotelling T²
            T2 = np.sum(
                (Z[:, :k] ** 2) / eigvals[:k],
                axis=1
            )

            # Residual space
            Z_residual = Z[:, k:]

            # SPE / Q statistic
            Q = np.sum(
                Z_residual ** 2,
                axis=1
            )

            results[k] = {
                "Mean_T2": np.mean(T2),
                "Mean_Q": np.mean(Q)
            }

        return results
```

```python
engine = PCAOptimizationEngine(df_strain)

results = engine.compute_pca_metrics()

for k, metrics in results.items():

    print(f"\nSubspace Dimension k = {k}")

    print(f"Mean T² = {metrics['Mean_T2']:.4f}")

    print(f"Mean Q  = {metrics['Mean_Q']:.4f}")
```

## Numerical Verification Results

The dataset was first centered and the unbiased sample covariance matrix was computed using

$$
\mathbf{S}
=
\frac{1}{n-1}
\sum_{i=1}^{n}
(\mathbf{x}_i-\widehat{\boldsymbol{\mu}})
(\mathbf{x}_i-\widehat{\boldsymbol{\mu}})^T.
$$

The covariance matrix was decomposed using eigenvalue decomposition,

$$
\mathbf{S}
=
\mathbf{P}
\boldsymbol{\Lambda}
\mathbf{P}^T.
$$

For each subspace dimension \(k\), Hotelling's \(T^2\) statistic and the residual SPE statistic were calculated.

The Hotelling's statistic measures variation within the retained principal subspace,

$$
T_i^2
=
\sum_{j=1}^{k}
\frac{z_{ij}^2}{\lambda_j},
$$

while the SPE statistic measures the residual energy in the discarded subspace,

$$
Q_i
=
\sum_{j=k+1}^{m}
z_{ij}^2.
$$

As \(k\) increases, more variance is captured by the principal subspace, causing the average SPE value to decrease. This numerically verifies the orthogonal decomposition and variance conservation properties derived in Part A.

## 1. Fundamental Equation of Factor Analysis

Let

$$
\mathbf{Z}_i \in \mathbb{R}^{m}
$$

denote a standardized observation vector with

$$
E[\mathbf{Z}_i]=\mathbf{0}
$$

and correlation matrix

$$
\mathbf{R}
=
E[\mathbf{Z}_i\mathbf{Z}_i^T].
$$

Factor Analysis assumes that the observed variables are generated by a smaller set of latent factors according to

$$
\mathbf{Z}_i
=
\boldsymbol{\Lambda}\mathbf{F}_i
+
\boldsymbol{\varepsilon}_i,
$$

where

- \(\mathbf{F}_i \in \mathbb{R}^{k}\) is the latent factor vector,
- \(\boldsymbol{\Lambda}\in\mathbb{R}^{m\times k}\) is the factor loading matrix,
- \(\boldsymbol{\varepsilon}_i\in\mathbb{R}^{m}\) is the unique error vector.

The standard Factor Analysis assumptions are

$$
E[\mathbf{F}_i]=\mathbf{0},
$$

$$
Cov(\mathbf{F}_i)=\mathbf{I},
$$

$$
E[\boldsymbol{\varepsilon}_i]=\mathbf{0},
$$

$$
Cov(\boldsymbol{\varepsilon}_i)
=
\boldsymbol{\Psi}
=
diag(\varphi_1^2,\varphi_2^2,\ldots,\varphi_m^2),
$$

and

$$
Cov(\mathbf{F}_i,\boldsymbol{\varepsilon}_i)=\mathbf{0}.
$$

Taking the covariance of both sides,

$$
Cov(\mathbf{Z}_i)
=
Cov(\boldsymbol{\Lambda}\mathbf{F}_i+\boldsymbol{\varepsilon}_i).
$$

Expanding,

$$
\mathbf{R}
=
Cov(\boldsymbol{\Lambda}\mathbf{F}_i)
+
Cov(\boldsymbol{\varepsilon}_i).
$$

Since

$$
Cov(\mathbf{F}_i)=\mathbf{I},
$$

we obtain

$$
Cov(\boldsymbol{\Lambda}\mathbf{F}_i)
=
\boldsymbol{\Lambda}
Cov(\mathbf{F}_i)
\boldsymbol{\Lambda}^T
=
\boldsymbol{\Lambda}\boldsymbol{\Lambda}^T.
$$

Therefore,

$$
\boxed{
\mathbf{R}
=
\boldsymbol{\Lambda}
\boldsymbol{\Lambda}^T
+
\boldsymbol{\Psi}
}
$$

which is known as the **Fundamental Equation of Factor Analysis**.

The first term represents shared structural variation, while the second term represents variable-specific noise.

---

## 2. Communality and Uniqueness

Consider the \(j^{th}\) variable.

The diagonal entry of the correlation matrix can be written as

$$
R_{jj}
=
h_j^2+\varphi_j^2.
$$

Because the variables are standardized,

$$
R_{jj}=1.
$$

Hence,

$$
1=h_j^2+\varphi_j^2.
$$

### Communality

The communality of variable \(j\) is

$$
h_j^2
=
\sum_{r=1}^{k}
\lambda_{jr}^2.
$$

It measures the proportion of variance explained by the common latent factors.

A large communality indicates that the variable strongly participates in the shared structural behaviour of the system.


### Uniqueness

The uniqueness is

$$
\varphi_j^2
=
1-h_j^2.
$$

It represents the proportion of variance that cannot be explained by the common factors.

Large uniqueness values indicate that the variable contains mainly local noise, measurement error, or sensor-specific behaviour.

---

## 3. Effect of Varimax Rotation

The original loading matrix is often difficult to interpret because each variable may load moderately on several factors.

To improve interpretability, Factor Analysis applies an orthogonal rotation.

Let

$$
\mathbf{T}
$$

be an orthogonal rotation matrix satisfying

$$
\mathbf{T}^T\mathbf{T}
=
\mathbf{I}.
$$

The rotated loading matrix becomes

$$
\boldsymbol{\Lambda}_{rot}
=
\boldsymbol{\Lambda}\mathbf{T}.
$$

The covariance contribution after rotation is

$$
\boldsymbol{\Lambda}_{rot}
\boldsymbol{\Lambda}_{rot}^T
=
(\boldsymbol{\Lambda}\mathbf{T})
(\boldsymbol{\Lambda}\mathbf{T})^T.
$$

Expanding,

$$
=
\boldsymbol{\Lambda}
\mathbf{T}
\mathbf{T}^T
\boldsymbol{\Lambda}^T.
$$

Since

$$
\mathbf{T}\mathbf{T}^T
=
\mathbf{I},
$$

we obtain

$$
\boldsymbol{\Lambda}_{rot}
\boldsymbol{\Lambda}_{rot}^T
=
\boldsymbol{\Lambda}
\boldsymbol{\Lambda}^T.
$$

Therefore, Varimax rotation changes the orientation of the factors but preserves the covariance structure.

The advantage is that each variable tends to load strongly on one factor and weakly on the others, creating a simpler and more interpretable factor structure.

---

## 4. Thomson's Regression Method for Factor Scores

The latent factors cannot be observed directly.

To estimate them, we use Thomson's Regression Method.

Consider the joint vector

$$
\mathbf{Y}
=
\begin{bmatrix}
\mathbf{F}_i\\
\mathbf{Z}_i
\end{bmatrix}.
$$

Its covariance matrix is

$$
\boldsymbol{\Sigma}_{YY}
=
\begin{bmatrix}
\mathbf{I} & \boldsymbol{\Lambda}^T\\
\boldsymbol{\Lambda} & \mathbf{R}
\end{bmatrix}.
$$

For a multivariate normal distribution, the conditional expectation is

$$
E[\mathbf{F}_i|\mathbf{Z}_i]
=
Cov(\mathbf{F}_i,\mathbf{Z}_i)
Cov(\mathbf{Z}_i)^{-1}
\mathbf{Z}_i.
$$

Using

$$
Cov(\mathbf{F}_i,\mathbf{Z}_i)
=
\boldsymbol{\Lambda}^T
$$

and

$$
Cov(\mathbf{Z}_i)
=
\mathbf{R},
$$

the factor score estimator becomes

$$
\boxed{
\widehat{\mathbf{F}}_i
=
\boldsymbol{\Lambda}^T
\mathbf{R}^{-1}
\mathbf{Z}_i
}
$$

which is known as **Thomson's Regression Estimator**.

This estimator provides the minimum mean-square-error estimate of the hidden factor vector and is widely used in practical Factor Analysis implementations.


## Interpretation

Factor Analysis decomposes the total variance of a system into two components:

1. **Shared Structural Variance**
   $$\boldsymbol{\Lambda}\boldsymbol{\Lambda}^T$$

2. **Unique Sensor-Specific Variance**
   $$\boldsymbol{\Psi}$$

Communalities quantify how much of each variable is explained by common physical mechanisms, whereas uniqueness values quantify local noise and isolated sensor effects. Through Varimax rotation and Thomson's factor score estimation, the latent operational behaviour of the system can be interpreted and monitored efficiently.

## Data Standardization and Factor Analysis Engine

The following class implements the mathematical framework required for Factor Analysis.

The implementation performs:

- Z-score normalization
- Dimension validation
- Varimax rotated factor analysis
- Communality computation
- Uniqueness computation
- Thomson factor score estimation

```python
class FactorPartitionEngine:

    def __init__(self, df):
        self.df = df.copy()

    # --------------------------------------------------
    # Z-SCORE STANDARDIZATION
    # --------------------------------------------------

    def zscore_standardize(self):

        mean = self.df.mean()

        std = self.df.std(ddof=1)

        std[std == 0] = 1e-15

        Z = (self.df - mean) / std

        return Z

    # --------------------------------------------------
    # DIMENSION CHECKS
    # --------------------------------------------------

    def validate_dimensions(self, k):

        n = len(self.df)

        m = self.df.shape[1]

        if n <= m:

            raise ValueError(
                "Sample size must exceed feature dimension."
            )

        if k >= m:

            raise ValueError(
                "Number of latent factors must be less than number of features."
            )

    # --------------------------------------------------
    # FACTOR ANALYSIS
    # --------------------------------------------------

    def fit_factor_model(self, k):

        self.validate_dimensions(k)

        Z = self.zscore_standardize()

        fa = FactorAnalysis(
            n_components=k,
            rotation="varimax",
            random_state=42
        )

        factor_scores = fa.fit_transform(Z)

        return fa, factor_scores, Z

    # --------------------------------------------------
    # LOADINGS
    # --------------------------------------------------

    def get_loadings(self, fa):

        return fa.components_.T

    # --------------------------------------------------
    # COMMUNALITIES
    # --------------------------------------------------

    def compute_communalities(self, loadings):

        communalities = np.sum(
            loadings**2,
            axis=1
        )

        return communalities

    # --------------------------------------------------
    # UNIQUENESS
    # --------------------------------------------------

    def compute_uniqueness(self, communalities):

        uniqueness = 1 - communalities

        return uniqueness

    # --------------------------------------------------
    # MAIN ANALYSIS
    # --------------------------------------------------

    def analyze(self, k):

        fa, factor_scores, Z = self.fit_factor_model(k)

        loadings = self.get_loadings(fa)

        communalities = self.compute_communalities(loadings)

        uniqueness = self.compute_uniqueness(communalities)

        return {

            "fa": fa,

            "scores": factor_scores,

            "loadings": loadings,

            "communalities": communalities,

            "uniqueness": uniqueness,

            "columns": list(self.df.columns)
        }
```

# Question 4
# Part A: Theoretical Foundations of Subspace Diagnostics & Feature De-correlation

## 1. Principal Component Analysis versus Factor Analysis

Both Principal Component Analysis (PCA) and Factor Analysis (FA) are dimensionality reduction techniques, but they differ fundamentally in their objectives.

PCA seeks directions that maximize total variance in the observed data. Given a covariance matrix

$$
\mathbf{S},
$$

PCA performs the spectral decomposition

$$
\mathbf{S}
=
\mathbf{P}
\boldsymbol{\Lambda}
\mathbf{P}^{T},
$$

where

$$
\boldsymbol{\Lambda}
=
\text{diag}(\lambda_1,\lambda_2,\ldots,\lambda_m)
$$

contains the ordered eigenvalues and

$$
\mathbf{P}
=
[\mathbf{p}_1,\mathbf{p}_2,\ldots,\mathbf{p}_m]
$$

contains the corresponding eigenvectors.

The transformed coordinates are

$$
\mathbf{z}
=
\mathbf{P}^{T}\mathbf{x}.
$$

PCA therefore produces orthogonal principal components that capture the maximum amount of total variance.

In contrast, Factor Analysis assumes that the observed variables are generated by a smaller number of latent factors:

$$
\mathbf{x}
=
\boldsymbol{\Lambda}\mathbf{f}
+
\boldsymbol{\varepsilon}.
$$

The corresponding covariance model is

$$
\mathbf{R}
=
\boldsymbol{\Lambda}
\boldsymbol{\Lambda}^{T}
+
\boldsymbol{\Psi}.
$$

Unlike PCA, Factor Analysis separates shared structural variation from variable-specific noise.

Therefore:

- PCA focuses on variance maximization.
- FA focuses on discovering hidden physical factors.
- PCA does not distinguish signal from noise.
- FA explicitly partitions variance into common and unique components.

---

## 2. Variance Allocation and Feature De-correlation

The total variance of the PCA model is

$$
\text{tr}(\mathbf{S})
=
\sum_{j=1}^{m}\lambda_j.
$$

The cumulative explained variance ratio for retaining \(k\) components is

$$
\Phi(k)
=
\frac{\sum_{j=1}^{k}\lambda_j}
{\sum_{j=1}^{m}\lambda_j}.
$$

The residual unexplained variance is

$$
\Psi(k)
=
1-\Phi(k).
$$

As additional principal components are retained, the explained variance increases while the residual variance decreases.

Because PCA transforms the covariance matrix into

$$
Cov(\mathbf{z})
=
\boldsymbol{\Lambda},
$$

all cross-correlations are removed:

$$
Cov(z_i,z_j)=0,
\quad i\neq j.
$$

This process is known as feature de-correlation.

---

## 3. Hotelling's \(T^2\) Statistic

For a retained PCA subspace of dimension \(k\), Hotelling's statistic is

$$
T_i^2
=
\sum_{j=1}^{k}
\frac{z_{ij}^{2}}
{\lambda_j}.
$$

The statistic measures how far an observation lies from the center of the retained principal component space.

Large values of \(T^2\) indicate abnormal behaviour occurring within the dominant variance directions of the system.

Typical causes include:

- Operational load changes.
- Environmental variations.
- Global structural shifts.

Therefore, Hotelling's \(T^2\) monitors behaviour occurring inside the modeled PCA subspace.

---

## 4. Squared Prediction Error (Q Statistic)

The residual vector is

$$
\mathbf{e}_i
=
\mathbf{x}_i
-
\widehat{\mathbf{x}}_i.
$$

The Squared Prediction Error (SPE) statistic is

$$
Q_i
=
\|\mathbf{e}_i\|^{2}.
$$

This statistic measures energy that remains outside the retained PCA subspace.

Large SPE values indicate that the observation contains information not represented by the learned model.

Typical causes include:

- Sensor faults.
- Localized structural damage.
- Previously unseen operating conditions.

Consequently, the Q statistic acts as a detector for residual anomalies.

---

## 5. Communality and Uniqueness in Factor Analysis

For the \(j^{th}\) variable, the communality is

$$
h_j^{2}
=
\sum_{r=1}^{k}
\lambda_{jr}^{2}.
$$

Communality represents the proportion of variance explained by the common latent factors.

The uniqueness is

$$
\varphi_j^{2}
=
1-h_j^{2}.
$$

Uniqueness measures variance that is specific to a single variable and cannot be explained by the common factors.

Interpretation:

- High communality indicates strong participation in the shared structural behaviour.
- High uniqueness indicates localized noise or isolated sensor behaviour.

Thus, Factor Analysis provides a physically interpretable decomposition of system variance.

---

## 6. Diagnostic Interpretation

PCA and Factor Analysis provide complementary diagnostic information.

- PCA monitors global variance structure through principal components.
- Hotelling's \(T^2\) detects abnormalities within the retained subspace.
- SPE detects abnormalities outside the retained subspace.
- Factor Analysis separates shared structural effects from isolated noise sources.
- Communality identifies sensors that strongly contribute to common system behaviour.
- Uniqueness identifies sensors dominated by local disturbances.

Therefore, combining PCA and Factor Analysis provides a comprehensive framework for subspace diagnostics, feature de-correlation, anomaly detection, and structural health monitoring.

```python
class PCAOptimizationEngine:

    def __init__(self, df):

        self.df = df.copy()
```

```python
class PCAOptimizationEngine:

    def __init__(self, df):

        self.df = df.copy()

    def compute_empirical_pca(self):

        # ------------------------------
        # Extract numeric data
        # ------------------------------

        X = self.df.values

        n, m = X.shape

        # ------------------------------
        # Mean center
        # ------------------------------

        mu_hat = np.mean(X, axis=0)

        X_centered = X - mu_hat

        # ------------------------------
        # Covariance matrix
        # ------------------------------

        S = np.cov(
            X_centered,
            rowvar=False,
            ddof=1
        )

        # ------------------------------
        # Eigen decomposition
        # ------------------------------

        eigvals, eigvecs = np.linalg.eigh(S)

        idx = np.argsort(eigvals)[::-1]

        eigvals = eigvals[idx]

        eigvecs = eigvecs[:, idx]

        # ------------------------------
        # Explained variance
        # ------------------------------

        explained = eigvals / np.sum(eigvals)

        cumulative = np.cumsum(explained)

        # ------------------------------
        # PCA scores
        # ------------------------------

        Z = X_centered @ eigvecs

        # ------------------------------
        # T² and Q profiles
        # ------------------------------

        k_values = np.arange(1, m + 1)

        mean_T2 = []

        mean_Q = []

        for k in k_values:

            Zk = Z[:, :k]

            lambdak = eigvals[:k]

            T2 = np.sum(
                (Zk**2) / lambdak,
                axis=1
            )

            mean_T2.append(np.mean(T2))

            residual = Z[:, k:]

            Q = np.sum(
                residual**2,
                axis=1
            )

            mean_Q.append(np.mean(Q))

        return {

            "eigenvectors": eigvecs,

            "eigenvalues": eigvals,

            "explained_variance_ratio": explained,

            "cumulative_variance_ratio": cumulative,

            "mean_T2_profile": np.array(mean_T2),

            "mean_Q_profile": np.array(mean_Q),

            "features": list(self.df.columns),

            "k_values": k_values
        }
```

```python
engine = PCAOptimizationEngine(df_asset)
```

```python
# Run the PCA engine created in the previous section
# This computes eigenvalues, eigenvectors,
# explained variance, T² statistics and Q statistics

pca_results = engine.compute_empirical_pca()
```
