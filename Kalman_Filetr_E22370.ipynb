## Q1. Analytical Derivation

We are given the prediction model

$$
x_k^- = A_{k-1}x_{k-1}^+ + G_{k-1}w_{k-1},
$$

where

$$
x_{k-1}^+ \sim \mathcal N(m_{k-1},P_{k-1}),
$$

and the process noise

$$
w_{k-1}\sim\mathcal N(0,\Sigma_p),
$$

is independent of the previous state estimate.

Since the state equation is a linear transformation of Gaussian random variables, the predicted state is also Gaussian. Hence,

$$
x_k^- \sim \mathcal N(m_k^-,P_k^-).
$$

### Step 1: Derivation of the Predicted Mean

Taking expectation on both sides of the state equation,

$$
\mathbb E[x_k^-]
=
\mathbb E[A_{k-1}x_{k-1}^+ + G_{k-1}w_{k-1}].
$$

Using the linearity of expectation,

$$
=
A_{k-1}\mathbb E[x_{k-1}^+]
+
G_{k-1}\mathbb E[w_{k-1}].
$$

Since

$$
\mathbb E[x_{k-1}^+]=m_{k-1},
$$

and

$$
\mathbb E[w_{k-1}]=0,
$$

we obtain

$$
m_k^-
=
A_{k-1}m_{k-1}.
$$

### Step 2: Derivation of the Predicted Covariance

The covariance of the predicted state is

$$
P_k^-
=
\operatorname{Var}(x_k^-).
$$

Substituting the state equation,

$$
P_k^-
=
\operatorname{Var}
(A_{k-1}x_{k-1}^+
+
G_{k-1}w_{k-1}).
$$

Since the previous state estimate and process noise are mutually independent,

$$
\operatorname{Cov}(x_{k-1}^+,w_{k-1})=0.
$$

Therefore,

$$
P_k^-
=
A_{k-1}
\operatorname{Var}(x_{k-1}^+)
A_{k-1}^T
+
G_{k-1}
\operatorname{Var}(w_{k-1})
G_{k-1}^T.
$$

Using

$$
\operatorname{Var}(x_{k-1}^+)=P_{k-1},
$$

and

$$
\operatorname{Var}(w_{k-1})=\Sigma_p,
$$

gives

$$
P_k^-
=
A_{k-1}P_{k-1}A_{k-1}^T
+
G_{k-1}\Sigma_pG_{k-1}^T.
$$


Hence, the predicted state is Gaussian,

$$
{
x_k^-
\sim
\mathcal N(m_k^-,P_k^-)
}
$$

where

$$
{
m_k^-
=
A_{k-1}m_{k-1}
}
$$

and

$$
{
P_k^-
=
A_{k-1}P_{k-1}A_{k-1}^T
+
G_{k-1}\Sigma_pG_{k-1}^T.
}
$$

Thus, the required result has been proved.

##2. Measurement Prediction

We are given the measurement model

$$
y_k^- = H_kx_k^- + z_k,
$$

where

$$
x_k^- \sim \mathcal N(m_k^-,P_k^-),
$$

and

$$
z_k \sim \mathcal N(0,\Sigma_m),
$$

with the measurement noise being independent of the predicted state.

Since the measurement equation is a linear transformation of Gaussian random variables, the predicted measurement is also Gaussian. Therefore,

$$
y_k^- \sim \mathcal N(\mu_y,\Sigma_y).
$$

### Step 1: Derivation of the Predicted Measurement Mean

Taking expectation on both sides of the measurement equation,

$$
\mathbb E[y_k^-]
=
\mathbb E[H_kx_k^-+z_k].
$$

Using the linearity of expectation,

$$
=
H_k\mathbb E[x_k^-]
+
\mathbb E[z_k].
$$

Since

$$
\mathbb E[x_k^-]=m_k^-,
$$

and

$$
\mathbb E[z_k]=0,
$$

we obtain

$$
{
\mathbb E[y_k^-]
=
H_km_k^-.
}
$$

Thus, the predicted measurement mean is

$$
{
\mu_y=H_km_k^-.
}
$$

### Step 2: Derivation of the Predicted Measurement Covariance

The covariance of the predicted measurement is

$$
\Sigma_y
=
\operatorname{Var}(y_k^-).
$$

Substituting the measurement equation,

$$
\Sigma_y
=
\operatorname{Var}(H_kx_k^-+z_k).
$$

Since the predicted state and measurement noise are mutually independent,

$$
\operatorname{Cov}(x_k^-,z_k)=0.
$$

Hence,

$$
\Sigma_y
=
H_k\operatorname{Var}(x_k^-)H_k^T
+
\operatorname{Var}(z_k).
$$

Using

$$
\operatorname{Var}(x_k^-)=P_k^-,
$$

and

$$
\operatorname{Var}(z_k)=\Sigma_m,
$$

gives

$$
{
\Sigma_y
=
H_kP_k^-H_k^T+\Sigma_m.
}
$$

Therefore, the predicted measurement follows the Gaussian distribution

$$
{
y_k^-
\sim
\mathcal N
\left(
H_km_k^-,
\;
H_kP_k^-H_k^T+\Sigma_m
\right).
}
$$

Hence, it has been shown that

$$
{
y_k^-
\sim
\mathcal N(H_km_k^-,\,H_kP_k^-H_k^T+\Sigma_m).
}
$$

Thus, the required result is proved.

##3. Joint Prior Distribution

We have already shown that

$$
x_k^- \sim \mathcal N(m_k^-,P_k^-),
$$

and the measurement model is

$$
y_k^- = H_kx_k^- + z_k,
$$

where

$$
z_k\sim\mathcal N(0,\Sigma_m),
$$

and the measurement noise is independent of the predicted state.

### Step 1: Mean Vector

The mean of the predicted state is

$$
\mathbb E[x_k^-]=m_k^-.
$$

From Question 2, the mean of the predicted measurement is

$$
\mathbb E[y_k^-]=H_km_k^-.
$$

Therefore, the joint mean vector becomes

$$
{
\mathbb E
\left[
\begin{bmatrix}
x_k^-\\
y_k^-
\end{bmatrix}
\right]
=
\begin{bmatrix}
m_k^-\\
H_km_k^-
\end{bmatrix}.
}
$$

### Step 2: Covariance of the Predicted State

The covariance of the first random variable is

$$
\operatorname{Var}(x_k^-)=P_k^-.
$$

### Step 3: Covariance of the Predicted Measurement

From Question 2,

$$
\operatorname{Var}(y_k^-)
=
H_kP_k^-H_k^T+\Sigma_m.
$$

### Step 4: Cross-Covariance

The cross-covariance between the predicted state and predicted measurement is

$$
\operatorname{Cov}(x_k^-,y_k^-)
=
\operatorname{Cov}(x_k^-,H_kx_k^-+z_k).
$$

Using the linearity of covariance,

$$
=
\operatorname{Cov}(x_k^-,H_kx_k^-)
+
\operatorname{Cov}(x_k^-,z_k).
$$

Since the measurement noise is independent of the predicted state,

$$
\operatorname{Cov}(x_k^-,z_k)=0.
$$

Hence,

$$
\operatorname{Cov}(x_k^-,y_k^-)
=
\operatorname{Cov}(x_k^-,H_kx_k^-).
$$

Using the covariance identity

$$
\operatorname{Cov}(X,AY)=\operatorname{Cov}(X,Y)A^T,
$$

we obtain

$$
=
P_k^-H_k^T.
$$

Similarly,

$$
\operatorname{Cov}(y_k^-,x_k^-)
=
(H_kP_k^-H_k^T)^T
=
H_kP_k^-,
$$

since the covariance matrix is symmetric.

Therefore,

$$
{
\operatorname{Cov}(x_k^-,y_k^-)
=
P_k^-H_k^T,
}
$$

and

$$
{
\operatorname{Cov}(y_k^-,x_k^-)
=
H_kP_k^-.
}
$$

### Step 5: Form the Joint Covariance Matrix

Combining all covariance terms,

$$
\operatorname{Cov}
\left(
\begin{bmatrix}
x_k^-\\
y_k^-
\end{bmatrix}
\right)
=
\begin{bmatrix}
\operatorname{Var}(x_k^-)
&
\operatorname{Cov}(x_k^-,y_k^-)
\\[2mm]
\operatorname{Cov}(y_k^-,x_k^-)
&
\operatorname{Var}(y_k^-)
\end{bmatrix}.
$$

Substituting the derived expressions,

$$
=
\begin{bmatrix}
P_k^-
&
P_k^-H_k^T
\\[2mm]
H_kP_k^-
&
H_kP_k^-H_k^T+\Sigma_m
\end{bmatrix}.
$$

Therefore, the predicted state and predicted measurement form the following joint Gaussian distribution:

$$
{
\begin{bmatrix}
x_k^-\\
y_k^-
\end{bmatrix}
\sim
\mathcal N
\left(
\begin{bmatrix}
m_k^-\\
H_km_k^-
\end{bmatrix},
\begin{bmatrix}
P_k^-
&
P_k^-H_k^T
\\[2mm]
H_kP_k^-
&
H_kP_k^-H_k^T+\Sigma_m
\end{bmatrix}
\right).
}
$$

Thus, the required joint prior distribution has been proved.

##4. Measurement Update (Correction Step)

From Question 3, the predicted state and predicted measurement have the joint Gaussian distribution

$$
\begin{bmatrix}
x_k^-\\
y_k^-
\end{bmatrix}
\sim
\mathcal N
\left(
\begin{bmatrix}
m_k^-\\
H_km_k^-
\end{bmatrix},
\begin{bmatrix}
P_k^- & P_k^-H_k^T\\
H_kP_k^- & H_kP_k^-H_k^T+\Sigma_m
\end{bmatrix}
\right).
$$

At time step \(k\), the actual measurement is observed as

$$
y_k = y_k^{\mathrm{obs}}.
$$

Since the joint distribution is Gaussian, the posterior distribution is obtained by conditioning on the observed measurement.

Therefore,

$$
x_k^+
\triangleq
(x_k^-\,|\,y_k=y_k^{\mathrm{obs}})
\sim
\mathcal N(m_k,P_k).
$$

### Step 1: Derivation of the Kalman Gain

For a joint Gaussian distribution,

$$
\begin{bmatrix}
x\\
y
\end{bmatrix}
\sim
\mathcal N
\left(
\begin{bmatrix}
\mu_x\\
\mu_y
\end{bmatrix},
\begin{bmatrix}
\Sigma_{xx} & \Sigma_{xy}\\
\Sigma_{yx} & \Sigma_{yy}
\end{bmatrix}
\right),
$$

the conditional mean is

$$
\mu_{x|y}
=
\mu_x
+
\Sigma_{xy}
\Sigma_{yy}^{-1}
(y-\mu_y).
$$

Comparing with the Kalman Filter matrices,

$$
\Sigma_{xy}=P_k^-H_k^T,
$$

and

$$
\Sigma_{yy}
=
H_kP_k^-H_k^T+\Sigma_m.
$$

Hence, the Kalman Gain is

$$
{
K_k
=
P_k^-H_k^T
\left(
H_kP_k^-H_k^T+\Sigma_m
\right)^{-1}.
}
$$

### Step 2: Derivation of the Updated Mean

Applying the conditional mean formula,

$$
m_k
=
m_k^-
+
K_k
\left(
y_k^{\mathrm{obs}}
-
H_km_k^-
\right).
$$

The quantity

$$
y_k^{\mathrm{obs}}-H_km_k^-
$$

is called the **innovation (measurement residual)**.

Therefore,

$$
{
m_k
=
m_k^-
+
K_k
\left(
y_k^{\mathrm{obs}}
-
H_km_k^-
\right).
}
$$

### Step 3: Derivation of the Updated Covariance

For a conditional multivariate Gaussian,

$$
P_k
=
P_k^-
-
P_k^-H_k^T
\left(
H_kP_k^-H_k^T+\Sigma_m
\right)^{-1}
H_kP_k^-.
$$

Substituting the Kalman Gain,

$$
K_k
=
P_k^-H_k^T
\left(
H_kP_k^-H_k^T+\Sigma_m
\right)^{-1},
$$

gives

$$
P_k
=
P_k^-
-
K_kH_kP_k^-.
$$

Factoring out \(P_k^-\),

$$
{
P_k
=
(I-K_kH_k)P_k^-.
}
$$

Hence, the posterior state estimate is

$$
{
x_k^+
=
(x_k^-\,|\,y_k=y_k^{\mathrm{obs}})
\sim
\mathcal N(m_k,P_k).
}
$$

where the Kalman Gain is

$$
{
K_k
=
P_k^-H_k^T
\left(
H_kP_k^-H_k^T+\Sigma_m
\right)^{-1},
}
$$

the updated mean is

$$
{
m_k
=
m_k^-
+
K_k
\left(
y_k^{\mathrm{obs}}
-
H_km_k^-
\right),
}
$$

and the updated covariance is

$$
{
P_k
=
(I-K_kH_k)P_k^-.
}
$$

Thus, the required measurement update equations have been proved.

## 5. Error Dynamics

Define the estimation error as

$$
e_k \triangleq x_k^+ - m_k,
$$

where \(x_k^+\) is the updated state estimate and \(m_k\) is the posterior mean.

From the state equation,

$$
x_k^- = A_{k-1}x_{k-1}^+ + G_{k-1}w_{k-1},
$$

the prediction error is

$$
x_k^- - m_k^-
=
A_{k-1}(x_{k-1}^+ - m_{k-1})
+
G_{k-1}w_{k-1}.
$$

Since

$$
e_{k-1}=x_{k-1}^+-m_{k-1},
$$

we obtain

$$
x_k^- - m_k^-
=
A_{k-1}e_{k-1}
+
G_{k-1}w_{k-1}.
$$


### Measurement Update

The posterior mean is

$$
m_k
=
m_k^-
+
K_k
\left(
y_k^{\mathrm{obs}}
-
H_km_k^-
\right).
$$

The measurement equation is

$$
y_k^{\mathrm{obs}}
=
H_kx_k^-+z_k.
$$

Substituting into the innovation,

$$
y_k^{\mathrm{obs}}
-
H_km_k^-
=
H_k(x_k^- - m_k^-)
+
z_k.
$$

Hence,

$$
m_k
=
m_k^-
+
K_kH_k(x_k^- - m_k^-)
+
K_kz_k.
$$

### Estimation Error

The posterior estimation error is

$$
e_k
=
x_k^+
-
m_k.
$$

Substituting the updated mean,

$$
e_k
=
(x_k^- - m_k^-)
-
K_kH_k(x_k^- - m_k^-)
-
K_kz_k.
$$

Factoring,

$$
e_k
=
(I-K_kH_k)(x_k^- - m_k^-)
-
K_kz_k.
$$

Using

$$
x_k^- - m_k^-
=
A_{k-1}e_{k-1}
+
G_{k-1}w_{k-1},
$$

gives

$$
e_k
=
(I-K_kH_k)
\left(
A_{k-1}e_{k-1}
+
G_{k-1}w_{k-1}
\right)
-
K_kz_k.
$$

Expanding,

$$
{
e_k
=
(I-K_kH_k)A_{k-1}e_{k-1}
+
(I-K_kH_k)G_{k-1}w_{k-1}
-
K_kz_k.
}
$$

### Error Covariance

Taking the covariance of the error dynamics and using the independence of
\(e_{k-1}\), \(w_{k-1}\), and \(z_k\),

$$
{
P_k
=
(I-K_kH_k)
\left(
A_{k-1}P_{k-1}A_{k-1}^T
+
G_{k-1}\Sigma_pG_{k-1}^T
\right).
}
$$

Since

$$
P_k^-
=
A_{k-1}P_{k-1}A_{k-1}^T
+
G_{k-1}\Sigma_pG_{k-1}^T,
$$

the covariance update can be written as

$$
{
P_k=(I-K_kH_k)P_k^-.
}
$$

Therefore, the Kalman Filter error dynamics are

$$
{
e_k
=
(I-K_kH_k)A_{k-1}e_{k-1}
+
(I-K_kH_k)G_{k-1}w_{k-1}
-
K_kz_k,
}
$$

with the corresponding error covariance

$$
{
P_k=(I-K_kH_k)P_k^-.
}
$$

Thus, the recursive error propagation equation of the Kalman Filter has been derived.

## Q1-D Example

1.Consider the scalar linear-Gaussian filter model

$$
x_k^- = ax_{k-1}^+ + w_{k-1},
$$

where

$$
x_{k-1}^+ \sim \mathcal N(m_{k-1},P_{k-1}),
$$

and

$$
w_{k-1}\sim\mathcal N(0,q),
$$

with the process noise independent of the previous state estimate.

Since the state equation is a linear transformation of Gaussian random variables, the predicted state is also Gaussian. Therefore,

$$
x_k^- \sim \mathcal N(m_k^-,P_k^-).
$$

### Step 1: Derivation of the Predicted Mean

Taking the expectation of the state equation,

$$
\mathbb E[x_k^-]
=
\mathbb E[ax_{k-1}^+ + w_{k-1}].
$$

Using the linearity of expectation,

$$
=
a\mathbb E[x_{k-1}^+]
+
\mathbb E[w_{k-1}].
$$

Since

$$
\mathbb E[x_{k-1}^+]=m_{k-1},
$$

and

$$
\mathbb E[w_{k-1}]=0,
$$

the predicted mean becomes

$$
{
m_k^-
=
am_{k-1}.
}
$$

### Step 2: Derivation of the Predicted Variance

The variance of the predicted state is

$$
P_k^-
=
\operatorname{Var}(x_k^-).
$$

Substituting the state equation,

$$
P_k^-
=
\operatorname{Var}
(ax_{k-1}^+ + w_{k-1}).
$$

Since the previous state estimate and the process noise are independent,

$$
\operatorname{Cov}(x_{k-1}^+,w_{k-1})=0.
$$

Hence,

$$
P_k^-
=
a^2\operatorname{Var}(x_{k-1}^+)
+
\operatorname{Var}(w_{k-1}).
$$

Using

$$
\operatorname{Var}(x_{k-1}^+)=P_{k-1},
$$

and

$$
\operatorname{Var}(w_{k-1})=q,
$$

gives

$$
{
P_k^-
=
a^2P_{k-1}+q.
}
$$

Therefore, the predicted state distribution is

$$
{
x_k^-
\sim
\mathcal N(m_k^-,P_k^-),
}
$$

where

$$
{
m_k^-
=
am_{k-1},
}
$$

and

$$
{
P_k^-
=
a^2P_{k-1}+q.
}
$$

Thus, the required result has been proved.

2.From Question 1, the predicted state is

$$
x_k^- \sim \mathcal N(m_k^-,P_k^-).
$$

The scalar measurement model is

$$
y_k^- = hx_k^- + z_k,
$$

where

$$
z_k\sim\mathcal N(0,r),
$$

and the measurement noise is independent of the predicted state.

From the prediction step, the predicted measurement distribution is

$$
y_k^- \sim \mathcal N(hm_k^-,S_k),
$$

where

$$
S_k=h^2P_k^-+r.
$$

### Step 1: Kalman Gain

For the scalar case,

$$
K_k
=
\frac{\operatorname{Cov}(x_k^-,y_k^-)}
{\operatorname{Var}(y_k^-)}.
$$

Since

$$
\operatorname{Cov}(x_k^-,y_k^-)
=
P_k^-h,
$$

and

$$
\operatorname{Var}(y_k^-)
=
S_k
=
h^2P_k^-+r,
$$

the Kalman Gain becomes

$$
{
K_k
=
\frac{P_k^-h}{S_k}.
}
$$

### Step 2: Innovation (Measurement Residual)

The innovation is defined as

$$
v_k
=
y_k^{\mathrm{obs}}
-
hm_k^-.
$$

It represents the difference between the observed measurement and the predicted measurement.

### Step 3: Updated Mean

The posterior mean is obtained by correcting the predicted mean using the innovation,

$$
m_k
=
m_k^-+K_kv_k.
$$

Substituting the expression for the innovation,

$$
m_k
=
m_k^-
+
K_k
\left(
y_k^{\mathrm{obs}}
-
hm_k^-
\right).
$$

Replacing \(K_k\),

$$
{
m_k
=
m_k^-
+
\frac{P_k^-h}{S_k}
\left(
y_k^{\mathrm{obs}}
-
hm_k^-
\right).
}
$$

### Step 4: Updated Variance

The posterior variance is

$$
P_k
=
(1-K_kh)P_k^-.
$$

Substituting the Kalman Gain,

$$
P_k
=
\left(
1-
\frac{P_k^-h}{S_k}h
\right)
P_k^-.
$$

Therefore,

$$
{
P_k
=
\left(
1-
\frac{P_k^-h^2}{S_k}
\right)
P_k^-.
}
$$


Hence, the scalar Kalman Filter measurement update equations are

$$
{
m_k
=
m_k^-+K_kv_k
=
m_k^-
+
\frac{P_k^-h}{S_k}
\left(
y_k^{\mathrm{obs}}
-
hm_k^-
\right),
}
$$

and

$$
{
P_k
=
(1-K_kh)P_k^-
=
\left(
1-
\frac{P_k^-h^2}{S_k}
\right)
P_k^-.
}
$$

Thus, the required scalar measurement update equations have been proved.

3.The measurement equation of the scalar Kalman Filter is

$$
y_k^- = hx_k^- + z_k,
$$

where

$$
x_k^- \sim \mathcal N(m_k^-,P_k^-),
$$

and

$$
z_k \sim \mathcal N(0,r).
$$

The process noise \(z_k\) is independent of the predicted state \(x_k^-\).

Our objective is to derive the predictive measurement distribution before observing the measurement,

$$
p(y_k^-|Y_{k-1}).
$$

Since \(x_k^-\) is Gaussian and the measurement model is linear with additive Gaussian noise, the predicted measurement is also Gaussian.

### Step 1: Compute the Mean

Taking the conditional expectation,

$$
\mathbb E[y_k^-|Y_{k-1}]
=
\mathbb E[hx_k^-+z_k|Y_{k-1}].
$$

Using the linearity of expectation,

$$
=
h\mathbb E[x_k^-|Y_{k-1}]
+
\mathbb E[z_k].
$$

Since

$$
\mathbb E[x_k^-|Y_{k-1}]
=
m_k^-,
$$

and

$$
\mathbb E[z_k]=0,
$$

we obtain

$$
{
\mathbb E[y_k^-|Y_{k-1}]
=
hm_k^-.
}
$$


### Step 2: Compute the Variance

The conditional variance is

$$
\operatorname{Var}(y_k^-|Y_{k-1})
=
\operatorname{Var}(hx_k^-+z_k).
$$

Since \(x_k^-\) and \(z_k\) are independent,

$$
\operatorname{Cov}(x_k^-,z_k)=0.
$$

Therefore,

$$
\operatorname{Var}(y_k^-|Y_{k-1})
=
h^2\operatorname{Var}(x_k^-)
+
\operatorname{Var}(z_k).
$$

Using

$$
\operatorname{Var}(x_k^-)=P_k^-,
$$

and

$$
\operatorname{Var}(z_k)=r,
$$

gives

$$
{
\operatorname{Var}(y_k^-|Y_{k-1})
=
h^2P_k^-+r.
}
$$

Define

$$
{
S_k=h^2P_k^-+r,
}
$$

which is called the **innovation covariance**.

### Step 3: Predictive Measurement Distribution

Combining the mean and variance,

$$
{
p(y_k^-|Y_{k-1})
=
\mathcal N
\left(
hm_k^-,
h^2P_k^-+r
\right).
}
$$

Equivalently,

$$
{
p(y_k^-|Y_{k-1})
=
\mathcal N(hm_k^-,S_k),
}
$$

where

$$
S_k=h^2P_k^-+r.
$$

Hence, the predictive measurement distribution before observing \(y_k\) is

$$
{
p(y_k^-|Y_{k-1})
=
\mathcal N(hm_k^-,\,h^2P_k^-+r).
}
$$

Thus, the required result has been proved.

---
4.From Question 2, after incorporating the measurement \(y_k^{\mathrm{obs}}\), the posterior state distribution is

$$
x_k^+|Y_k \sim \mathcal{N}(m_k,P_k).
$$

The measurement model is

$$
y_k^- = hx_k^+ + z_k,
$$

where

$$
z_k \sim \mathcal{N}(0,r),
$$

and the measurement noise is independent of the posterior state estimate.

We are required to derive the posterior-predictive measurement distribution

$$
p(y_k^-|Y_k).
$$

Since \(x_k^+|Y_k\) is Gaussian and the measurement model is linear with additive Gaussian noise, the posterior-predictive measurement is also Gaussian.

### Step 1: Derivation of the Mean

Taking the conditional expectation,

$$
\mathbb{E}[y_k^-|Y_k]
=
\mathbb{E}[hx_k^+ + z_k|Y_k].
$$

Using the linearity of expectation,

$$
=
h\mathbb{E}[x_k^+|Y_k]
+
\mathbb{E}[z_k].
$$

Since

$$
\mathbb{E}[x_k^+|Y_k]
=
m_k,
$$

and

$$
\mathbb{E}[z_k]=0,
$$

the posterior-predictive mean becomes

$$
\mathbb{E}[y_k^-|Y_k]
=
hm_k.
$$

### Step 2: Derivation of the Variance

The conditional variance is

$$
\operatorname{Var}(y_k^-|Y_k)
=
\operatorname{Var}(hx_k^+ + z_k).
$$

Since the posterior state estimate and the measurement noise are independent,

$$
\operatorname{Cov}(x_k^+,z_k)=0,
$$

therefore,

$$
\operatorname{Var}(y_k^-|Y_k)
=
h^2\operatorname{Var}(x_k^+)
+
\operatorname{Var}(z_k).
$$

Using

$$
\operatorname{Var}(x_k^+)=P_k,
$$

and

$$
\operatorname{Var}(z_k)=r,
$$

gives

$$
\operatorname{Var}(y_k^-|Y_k)
=
h^2P_k+r.
$$

### Step 3: Posterior-Predictive Measurement Distribution

Combining the mean and variance,

$$
p(y_k^-|Y_k)
=
\mathcal{N}
\left(
hm_k,\,
h^2P_k+r
\right).
$$

Hence, the posterior-predictive measurement distribution after filtering on the measurement \(y_k\) is

$$
p(y_k^-|Y_k)
=
\mathcal{N}(hm_k,\;h^2P_k+r).
$$

Thus, the required result has been proved.

---
5.From the prediction step,

$$
x_k^- \sim \mathcal{N}(m_k^-,P_k^-),
$$

and the measurement model is

$$
y_k^- = hx_k^- + z_k,
$$

where

$$
z_k \sim \mathcal{N}(0,r),
$$

with \(z_k\) independent of \(x_k^-\).

From Question 3, the predicted measurement is

$$
y_k^- \sim \mathcal{N}(hm_k^-,S_k),
$$

where

$$
S_k=h^2P_k^-+r.
$$

We now derive the posterior state distribution after observing the measurement \(y_k^{obs}\).

### Step 1: Kalman Gain

The covariance between the predicted state and the predicted measurement is

$$
\operatorname{Cov}(x_k^-,y_k^-)
=
P_k^-h.
$$

The variance of the predicted measurement is

$$
\operatorname{Var}(y_k^-)
=
S_k
=
h^2P_k^-+r.
$$

Therefore, the Kalman Gain is

$$
K_k
=
\frac{P_k^-h}{S_k}.
$$

### Step 2: Posterior Mean

For a conditional Gaussian distribution,

$$
m_k
=
m_k^-
+
K_k
\left(
y_k^{obs}
-
hm_k^-
\right).
$$

Substituting the Kalman Gain,

$$
m_k
=
m_k^-
+
\frac{P_k^-h}{S_k}
\left(
y_k^{obs}
-
hm_k^-
\right).
$$

### Step 3: Posterior Variance

The covariance of the conditional distribution is

$$
P_k
=
P_k^-
-
K_khP_k^-.
$$

Substituting the Kalman Gain,

$$
P_k
=
P_k^-
-
\frac{P_k^-h}{S_k}
hP_k^-.
$$

Hence,

$$
P_k
=
\left(
1-
\frac{P_k^-h^2}{S_k}
\right)
P_k^-.
$$

### Step 4: Posterior Distribution

Combining the posterior mean and posterior variance,

$$
x_k^+|y_k^{obs}
\sim
\mathcal{N}(m_k,P_k),
$$

where

$$
m_k
=
m_k^-
+
\frac{P_k^-h}{S_k}
\left(
y_k^{obs}
-
hm_k^-
\right),
$$

and

$$
P_k
=
\left(
1-
\frac{P_k^-h^2}{S_k}
\right)
P_k^-.
$$

Thus, the posterior state estimate after incorporating the measurement has been derived.

## Q.2D Position Estimation – Part A

The hidden state of the system is

$$
x_k=
\begin{bmatrix}
p_x(k)\\
p_y(k)\\
v_x(k)\\
v_y(k)
\end{bmatrix},
$$

where \(p_x(k)\) and \(p_y(k)\) are the position components, while \(v_x(k)\) and \(v_y(k)\) are the velocity components.

The measurements consist only of the position coordinates,

$$
y_k=
\begin{bmatrix}
p_x^{\mathrm{meas}}(k)\\
p_y^{\mathrm{meas}}(k)
\end{bmatrix}.
$$

The state-space model is

$$
x_k^-=Ax_{k-1}^+ + Gw_{k-1},
$$

$$
y_k^-=Hx_k^-+z_k.
$$

The process noise \(w_k\), measurement noise \(z_k\), and the initial state are assumed to be mutually independent.

### Step 1: Derivation of the State Transition Matrix \(A\)

For a constant-velocity model,

$$
p_x(k)=p_x(k-1)+v_x(k-1)\Delta t,
$$

$$
p_y(k)=p_y(k-1)+v_y(k-1)\Delta t,
$$

Since the velocity is assumed constant,

$$
v_x(k)=v_x(k-1),
$$

$$
v_y(k)=v_y(k-1).
$$

Writing these four equations in matrix form,

$$
\begin{bmatrix}
p_x(k)\\
p_y(k)\\
v_x(k)\\
v_y(k)
\end{bmatrix}
=
\begin{bmatrix}
1&0&\Delta t&0\\
0&1&0&\Delta t\\
0&0&1&0\\
0&0&0&1
\end{bmatrix}
\begin{bmatrix}
p_x(k-1)\\
p_y(k-1)\\
v_x(k-1)\\
v_y(k-1)
\end{bmatrix}.
$$

Therefore,

$$
A=
\begin{bmatrix}
1&0&\Delta t&0\\
0&1&0&\Delta t\\
0&0&1&0\\
0&0&0&1
\end{bmatrix}.
$$

### Step 2: Derivation of the Measurement Matrix \(H\)

The measurements contain only the position coordinates.

Therefore,

$$
p_x^{\mathrm{meas}}=p_x,
$$

$$
p_y^{\mathrm{meas}}=p_y.
$$

The velocity components are not directly measured.

Hence,

$$
\begin{bmatrix}
p_x^{\mathrm{meas}}\\
p_y^{\mathrm{meas}}
\end{bmatrix}
=
\begin{bmatrix}
1&0&0&0\\
0&1&0&0
\end{bmatrix}
\begin{bmatrix}
p_x\\
p_y\\
v_x\\
v_y
\end{bmatrix}.
$$

Therefore,

$$
H=
\begin{bmatrix}
1&0&0&0\\
0&1&0&0
\end{bmatrix}.
$$

### Step 3: Derivation of the Process Noise Matrix \(G\)

Assume that the process noise is acceleration,

$$
w_k=
\begin{bmatrix}
a_x\\
a_y
\end{bmatrix}.
$$

Using the equations of uniformly accelerated motion,

$$
p=p_0+v_0\Delta t+\frac12a\Delta t^2,
$$

$$
v=v_0+a\Delta t.
$$

For the x-direction,

$$
\Delta x=\frac12a_x\Delta t^2,
$$

$$
\Delta v_x=a_x\Delta t.
$$

Similarly, for the y-direction,

$$
\Delta y=\frac12a_y\Delta t^2,
$$

$$
\Delta v_y=a_y\Delta t.
$$

Collecting these terms,

$$
\begin{bmatrix}
\Delta p_x\\
\Delta p_y\\
\Delta v_x\\
\Delta v_y
\end{bmatrix}
=
\begin{bmatrix}
\frac12\Delta t^2&0\\
0&\frac12\Delta t^2\\
\Delta t&0\\
0&\Delta t
\end{bmatrix}
\begin{bmatrix}
a_x\\
a_y
\end{bmatrix}.
$$

Hence,

$$
G=
\begin{bmatrix}
\frac12\Delta t^2&0\\
0&\frac12\Delta t^2\\
\Delta t&0\\
0&\Delta t
\end{bmatrix}.
$$

Thus, the matrices for the two-dimensional constant-velocity Kalman Filter model are

$$
A=
\begin{bmatrix}
1&0&\Delta t&0\\
0&1&0&\Delta t\\
0&0&1&0\\
0&0&0&1
\end{bmatrix},
$$

$$
H=
\begin{bmatrix}
1&0&0&0\\
0&1&0&0
\end{bmatrix},
$$

and

$$
G=
\begin{bmatrix}
\frac12\Delta t^2&0\\
0&\frac12\Delta t^2\\
\Delta t&0\\
0&\Delta t
\end{bmatrix}.
$$

Hence, the required state transition matrix, measurement matrix, and process noise matrix have been derived.

### Step 4: Derivation of the Process Noise Covariance

Assume that the process noise covariance is

$$
\Sigma_p=qI_2,
$$

where

$$
I_2=
\begin{bmatrix}
1&0\\
0&1
\end{bmatrix}.
$$

The induced state-space process covariance is

$$
Q=G\Sigma_pG^T.
$$

Substituting

$$
\Sigma_p=qI_2,
$$

gives

$$
Q=qGG^T.
$$

Using

$$
G=
\begin{bmatrix}
\frac12\Delta t^2&0\\
0&\frac12\Delta t^2\\
\Delta t&0\\
0&\Delta t
\end{bmatrix},
$$

its transpose is

$$
G^T=
\begin{bmatrix}
\frac12\Delta t^2&0&\Delta t&0\\
0&\frac12\Delta t^2&0&\Delta t
\end{bmatrix}.
$$

Therefore,

$$
Q
=
q
\begin{bmatrix}
\frac12\Delta t^2&0\\
0&\frac12\Delta t^2\\
\Delta t&0\\
0&\Delta t
\end{bmatrix}
\begin{bmatrix}
\frac12\Delta t^2&0&\Delta t&0\\
0&\frac12\Delta t^2&0&\Delta t
\end{bmatrix}.
$$

Performing the matrix multiplication,

$$
Q
=
q
\begin{bmatrix}
\frac14\Delta t^4&0&\frac12\Delta t^3&0\\
0&\frac14\Delta t^4&0&\frac12\Delta t^3\\
\frac12\Delta t^3&0&\Delta t^2&0\\
0&\frac12\Delta t^3&0&\Delta t^2
\end{bmatrix}.
$$

Thus, the complete Gaussian state-space model becomes

$$
x_k^-
=
Ax_{k-1}^+
+
Gw_{k-1},
\qquad
w_{k-1}\sim\mathcal N(0,qI_2),
$$

$$
y_k^-
=
Hx_k^-
+
z_k,
\qquad
z_k\sim\mathcal N(0,rI_2).
$$

where

$$
A=
\begin{bmatrix}
1&0&\Delta t&0\\
0&1&0&\Delta t\\
0&0&1&0\\
0&0&0&1
\end{bmatrix},
$$

$$
H=
\begin{bmatrix}
1&0&0&0\\
0&1&0&0
\end{bmatrix},
$$

$$
G=
\begin{bmatrix}
\frac12\Delta t^2&0\\
0&\frac12\Delta t^2\\
\Delta t&0\\
0&\Delta t
\end{bmatrix},
$$

and

$$
Q
=
q
\begin{bmatrix}
\frac14\Delta t^4&0&\frac12\Delta t^3&0\\
0&\frac14\Delta t^4&0&\frac12\Delta t^3\\
\frac12\Delta t^3&0&\Delta t^2&0\\
0&\frac12\Delta t^3&0&\Delta t^2
\end{bmatrix}.
$$

Hence, the required matrices \(A\), \(H\), \(G\), and the induced process covariance matrix \(Q\) have been successfully derived.

# Part B: Python-Based Kalman Filter for 2D Position Estimation

## Methodology

A Kalman Filter is developed to estimate the true position of an object from noisy GPS measurements using the constant-velocity motion model derived in Part A.

The state vector is defined as

$$
x_k=
\begin{bmatrix}
p_x(k)\\
p_y(k)\\
v_x(k)\\
v_y(k)
\end{bmatrix},
$$

where \(p_x(k)\) and \(p_y(k)\) represent the position coordinates, while \(v_x(k)\) and \(v_y(k)\) represent the corresponding velocities.

The measurement vector contains only the observed GPS positions,

$$
y_k=
\begin{bmatrix}
p_x^{meas}(k)\\
p_y^{meas}(k)
\end{bmatrix}.
$$

The Kalman Filter operates recursively using two stages.

### Prediction Step

The state estimate is predicted using

$$
x_k^- = Ax_{k-1}^+,
$$

and the covariance is propagated as

$$
P_k^- = AP_{k-1}A^T + Q,
$$

where

$$
Q=G\Sigma_pG^T.
$$

### Measurement Update

After receiving a new GPS measurement, the innovation covariance is computed as

$$
S_k = HP_k^-H^T + R.
$$

The Kalman Gain is

$$
K_k=P_k^-H^TS_k^{-1}.
$$

The updated state estimate becomes

$$
x_k^+
=
x_k^-
+
K_k
(y_k-Hx_k^-),
$$

while the updated covariance is

$$
P_k
=
(I-K_kH)P_k^-.
$$

These prediction and correction steps are repeated for every GPS measurement in the sequence, allowing the filter to reduce measurement noise while providing smooth estimates of both position and velocity.

The implementation is carried out in Python using NumPy for matrix computations and Matplotlib for visualization of the noisy GPS measurements and the filtered trajectory.

```python
# ==========================================================
# 2D Constant Velocity Kalman Filter
# Part 1: Import libraries and generate the true trajectory
# ==========================================================

# Numerical computations
import numpy as np

# Plotting
import matplotlib.pyplot as plt

# Make the random numbers reproducible
np.random.seed(42)
```

```python
# ==========================================================
# Simulation Parameters
# ==========================================================

# Sampling time (seconds)
dt = 1.0

# Number of time steps
N = 100

# True constant velocities
vx_true = 2.0
vy_true = 1.0
```

```python
# ==========================================================
# Generate the true object trajectory
# ==========================================================

# Time vector
time = np.arange(N)

# True x-position
px_true = vx_true * time

# True y-position
py_true = vy_true * time
```

```python
# ==========================================================
# Plot the true trajectory
# ==========================================================

plt.figure(figsize=(7,6))

plt.plot(px_true,
         py_true,
         linewidth=3,
         label="True Trajectory")

plt.xlabel("X Position (m)")
plt.ylabel("Y Position (m)")

plt.title("True Object Trajectory")

plt.legend()

plt.grid(True)

plt.axis("equal")

plt.show()
```

```python
# ==========================================================
# Generate Noisy GPS Measurements
# ==========================================================

# Standard deviation of GPS measurement noise (meters)
gps_std = 5.0

# Generate Gaussian noise for x-coordinate
noise_x = np.random.normal(
    loc=0,
    scale=gps_std,
    size=N
)

# Generate Gaussian noise for y-coordinate
noise_y = np.random.normal(
    loc=0,
    scale=gps_std,
    size=N
)

# Add noise to the true positions
px_measured = px_true + noise_x
py_measured = py_true + noise_y
```

```python
# ==========================================================
# Plot the True Trajectory and Noisy GPS Measurements
# ==========================================================

plt.figure(figsize=(8,6))

# Plot true trajectory
plt.plot(
    px_true,
    py_true,
    'b-',
    linewidth=2,
    label="True Trajectory"
)

# Plot noisy GPS measurements
plt.scatter(
    px_measured,
    py_measured,
    color='red',
    s=35,
    alpha=0.7,
    label="Noisy GPS Measurements"
)

plt.title("True Trajectory vs Noisy GPS Measurements")

plt.xlabel("X Position (m)")
plt.ylabel("Y Position (m)")

plt.legend()

plt.grid(True)

plt.axis("equal")

plt.show()
```

```python
# ==========================================================
# Kalman Filter Model Matrices
# ==========================================================

# -----------------------------
# State Transition Matrix (A)
# -----------------------------
A = np.array([
    [1, 0, dt, 0],
    [0, 1, 0, dt],
    [0, 0, 1, 0],
    [0, 0, 0, 1]
])

# -----------------------------
# Measurement Matrix (H)
# -----------------------------
H = np.array([
    [1, 0, 0, 0],
    [0, 1, 0, 0]
])

# -----------------------------
# Process Noise Input Matrix (G)
# -----------------------------
G = np.array([
    [0.5 * dt**2, 0],
    [0, 0.5 * dt**2],
    [dt, 0],
    [0, dt]
])

# Display matrices
print("State Transition Matrix A:\n")
print(A)

print("\nMeasurement Matrix H:\n")
print(H)

print("\nProcess Noise Matrix G:\n")
print(G)
```

```python
# ==========================================================
# Noise Covariance Matrices
# ==========================================================

# Process noise variance (acceleration variance)
q = 0.1

# Process noise covariance
Sigma_p = q * np.eye(2)

# State process covariance
Q = G @ Sigma_p @ G.T

# Measurement noise variance (GPS variance)
r = gps_std**2

# Measurement covariance matrix
R = r * np.eye(2)

print("Process Noise Covariance Q:\n")
print(Q)

print("\nMeasurement Covariance R:\n")
print(R)
```

```python
# ==========================================================
# Initial State Estimate
# ==========================================================

# Initial estimate of the state vector
x = np.array([
    [0.0],   # Initial x-position
    [0.0],   # Initial y-position
    [0.0],   # Initial x-velocity
    [0.0]    # Initial y-velocity
])

# Initial covariance matrix
P = np.eye(4) * 100

print("Initial State Estimate:\n")
print(x)

print("\nInitial Covariance Matrix:\n")
print(P)
```

```python
# ==========================================================
# Kalman Filter Implementation
# ==========================================================

# Store estimated positions
estimated_x = []
estimated_y = []

# Identity matrix (used in covariance update)
I = np.eye(4)

# Loop through every GPS measurement
for k in range(N):

    # ==========================================
    # Prediction Step
    # ==========================================

    # Predict the next state
    x = A @ x

    # Predict the covariance
    P = A @ P @ A.T + Q


    # ==========================================
    # Measurement Update
    # ==========================================

    # Current GPS measurement
    z = np.array([
        [px_measured[k]],
        [py_measured[k]]
    ])

    # Innovation (measurement residual)
    y = z - H @ x

    # Innovation covariance
    S = H @ P @ H.T + R

    # Kalman Gain
    K = P @ H.T @ np.linalg.inv(S)

    # Correct the predicted state
    x = x + K @ y

    # Update covariance
    P = (I - K @ H) @ P

    # Save estimated position
    estimated_x.append(x[0,0])
    estimated_y.append(x[1,0])
```

```python
# ==========================================================
# Plot the Kalman Filter Results
# ==========================================================

plt.figure(figsize=(10, 8))

# Plot the true trajectory
plt.plot(
    px_true,
    py_true,
    color='blue',
    linewidth=2,
    label='True Trajectory'
)

# Plot the noisy GPS measurements
plt.scatter(
    px_measured,
    py_measured,
    color='red',
    s=30,
    alpha=0.6,
    label='Noisy GPS Measurements'
)

# Plot the Kalman Filter estimated trajectory
plt.plot(
    estimated_x,
    estimated_y,
    color='green',
    linewidth=2,
    label='Kalman Filter Estimate'
)

plt.title("2D Constant Velocity Kalman Filter")

plt.xlabel("X Position (m)")
plt.ylabel("Y Position (m)")

plt.legend()

plt.grid(True)

plt.axis("equal")

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(time, px_true, label="True X")
plt.scatter(time, px_measured, s=15, label="Measured X")
plt.plot(time, estimated_x, label="Estimated X")

plt.xlabel("Time (s)")
plt.ylabel("X Position (m)")
plt.title("X Position Estimation")

plt.grid(True)
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(time, py_true, label="True Y")
plt.scatter(time, py_measured, s=15, label="Measured Y")
plt.plot(time, estimated_y, label="Estimated Y")

plt.xlabel("Time (s)")
plt.ylabel("Y Position (m)")
plt.title("Y Position Estimation")

plt.grid(True)
plt.legend()
plt.show()
```

```python
gps_rmse = np.sqrt(
    np.mean(
        (px_true - px_measured)**2 +
        (py_true - py_measured)**2
    )
)

kf_rmse = np.sqrt(
    np.mean(
        (px_true - np.array(estimated_x))**2 +
        (py_true - np.array(estimated_y))**2
    )
)

print(f"GPS RMSE        : {gps_rmse:.3f} m")
print(f"Kalman RMSE     : {kf_rmse:.3f} m")
```
