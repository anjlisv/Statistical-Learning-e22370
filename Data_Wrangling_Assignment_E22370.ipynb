# Statistical Association Helper Functions

To generate a unified association heatmap, several statistical measures are required.

These include:

### Cramer's V
Used to measure associations between categorical variables.

### Eta Squared
Used to measure associations between categorical and numerical variables.

These helper functions will be used later when generating the comprehensive association heatmap.

```python
# ============================================================
# ASSOCIATION HELPER FUNCTIONS
# ============================================================

def cramers_v(x, y):

    confusion_matrix = pd.crosstab(x, y)

    chi2 = chi2_contingency(confusion_matrix)[0]

    n = confusion_matrix.sum().sum()

    phi2 = chi2 / n

    r, k = confusion_matrix.shape

    return np.sqrt(phi2 / min((k - 1), (r - 1)))


def eta_squared(categories, measurements):

    categories = pd.Categorical(categories)

    cat_num = categories.codes

    grand_mean = np.mean(measurements)

    ss_between = 0

    for category in np.unique(cat_num):

        vals = measurements[cat_num == category]

        ss_between += len(vals) * (np.mean(vals) - grand_mean)**2

    ss_total = np.sum((measurements - grand_mean)**2)

    return np.sqrt(ss_between / ss_total)
```

# DataInspector Class

The DataInspector class is the main component of the Data Sanitization and Exploration Engine.

The class is responsible for:

- Data ingestion
- Data sanitization
- Structural analysis
- Missing value handling
- Duplicate removal
- Outlier detection
- Feature engineering

- Interactive visualizations

This class works together with the PlottingMethods class to provide a complete data wrangling solution.

```python
# ============================================================
# DATA INSPECTOR CLASS
# ============================================================

class DataInspector:

    """
    Main data wrangling and exploration class.
    """

    def __init__(self):

        self.df = None

        self.plotter = PlottingMethods()

        print("DataInspector Initialized Successfully")
```

```python
# ============================================================
# DATA UPLOAD METHOD
# ============================================================

class DataInspector(DataInspector):

    def upload_data(self, file_path):

        """
        Upload and sanitize dataset.
        """

        self.df = pd.read_csv(file_path)

        garbage_strings = [
            '?',
            'n/a',
            'N/A',
            'NULL',
            '',
            ' '
        ]

        self.df.replace(
            garbage_strings,
            np.nan,
            inplace=True
        )

        for column in self.df.columns:

            try:

                converted = pd.to_numeric(
                    self.df[column]
                )

                if not converted.isna().all():

                    self.df[column] = converted

            except:

                pass

        print("Dataset Loaded Successfully")

        print(
            f"Rows : {self.df.shape[0]}"
        )

        print(
            f"Columns : {self.df.shape[1]}"
        )

        return self.df
```

```python
# ============================================================
# DATA SUMMARY METHOD
# ============================================================

class DataInspector(DataInspector):

    def data_summary(self):

        """
        Display structural information
        about the dataset.
        """

        print("=" * 60)

        print("DATA SUMMARY")

        print("=" * 60)

        print(f"\nRows : {self.df.shape[0]}")

        print(f"Columns : {self.df.shape[1]}")

        print("\nFIRST 20 RECORDS")

        display(self.df.head(20))

        numerical_columns = self.df.select_dtypes(
            include=np.number
        ).columns.tolist()

        categorical_columns = self.df.select_dtypes(
            exclude=np.number
        ).columns.tolist()

        print("\nNUMERICAL COLUMNS")

        print(numerical_columns)

        print("\nCATEGORICAL COLUMNS")

        print(categorical_columns)

        print("\nDATA TYPES")

        display(self.df.dtypes)
```

```python
# ============================================================
# HANDLE MISSING VALUES
# ============================================================

class DataInspector(DataInspector):

    def handle_missing_values(
        self,
        strategy="mean",
        constant_value="Unknown"
    ):

        """
        Missing value imputation.
        """

        for column in self.df.columns:

            missing_count = self.df[column].isna().sum()

            if missing_count == 0:

                continue

            if pd.api.types.is_numeric_dtype(
                self.df[column]
            ):

                if strategy == "mean":

                    self.df[column].fillna(
                        self.df[column].mean(),
                        inplace=True
                    )

                elif strategy == "median":

                    self.df[column].fillna(
                        self.df[column].median(),
                        inplace=True
                    )

                elif strategy == "mode":

                    self.df[column].fillna(
                        self.df[column].mode()[0],
                        inplace=True
                    )

                else:

                    self.df[column].fillna(
                        constant_value,
                        inplace=True
                    )

            else:

                if strategy == "mode":

                    self.df[column].fillna(
                        self.df[column].mode()[0],
                        inplace=True
                    )

                else:

                    self.df[column].fillna(
                        constant_value,
                        inplace=True
                    )

        print(
            f"Missing values handled using {strategy.upper()} strategy."
        )
```

```python
# ============================================================
# REMOVE DUPLICATES
# ============================================================

class DataInspector(DataInspector):

    def remove_duplicates(self):

        """
        Remove duplicate rows.
        """

        before_rows = len(self.df)

        self.df.drop_duplicates(
            inplace=True
        )

        after_rows = len(self.df)

        removed = before_rows - after_rows

        print(
            f"{removed} duplicate rows removed."
        )

        print(
            f"Current dataset size: {after_rows} rows."
        )
```

```python
# ============================================================
# HANDLE OUTLIERS USING IQR
# ============================================================

class DataInspector(DataInspector):

    def handle_outliers(
        self,
        columns,
        delete=False
    ):

        """
        Detect or remove outliers
        using the IQR method.
        """

        for column in columns:

            if not pd.api.types.is_numeric_dtype(
                self.df[column]
            ):
                continue

            Q1 = self.df[column].quantile(0.25)

            Q3 = self.df[column].quantile(0.75)

            IQR = Q3 - Q1

            lower_bound = Q1 - 1.5 * IQR

            upper_bound = Q3 + 1.5 * IQR

            outlier_mask = (
                (self.df[column] < lower_bound)
                |
                (self.df[column] > upper_bound)
            )

            outlier_count = outlier_mask.sum()

            print(
                f"{column}: {outlier_count} outliers detected"
            )

            if delete:

                self.df = self.df[
                    ~outlier_mask
                ]

                print(
                    f"Outliers removed from {column}"
                )
```

```python
# ============================================================
# DELETE ROWS
# ============================================================

class DataInspector(DataInspector):

    def delete_rows(self):

        """
        Delete rows by index.
        """

        rows = input(
            "Enter row indexes separated by commas: "
        )

        rows = [
            int(x.strip())
            for x in rows.split(",")
        ]

        self.df.drop(
            index=rows,
            inplace=True,
            errors="ignore"
        )

        print(
            "Selected rows deleted successfully."
        )
```

```python
# ============================================================
# DELETE COLUMNS
# ============================================================

class DataInspector(DataInspector):

    def delete_columns(self):

        """
        Delete columns by name.
        """

        columns = input(
            "Enter column names separated by commas: "
        )

        columns = [
            x.strip()
            for x in columns.split(",")
        ]

        self.df.drop(
            columns=columns,
            inplace=True,
            errors="ignore"
        )

        print(
            "Selected columns deleted successfully."
        )
```

```python
# ============================================================
# NUMERICAL NORMALIZATION
# ============================================================

class DataInspector(DataInspector):

    def extract_normalized_numeric_data(
        self,
        method="minmax"
    ):

        """
        Normalize numerical data using:
        - MinMax Scaling
        - Standard Scaling
        - Robust Scaling
        """

        numerical_data = self.df.select_dtypes(
            include=np.number
        )

        if method.lower() == "minmax":

            scaler = MinMaxScaler()

        elif method.lower() == "standard":

            scaler = StandardScaler()

        elif method.lower() == "robust":

            scaler = RobustScaler()

        else:

            raise ValueError(
                "Choose minmax, standard or robust"
            )

        transformed_data = scaler.fit_transform(
            numerical_data
        )

        normalized_df = pd.DataFrame(
            transformed_data,
            columns=numerical_data.columns
        )

        print(
            f"{method.upper()} scaling completed successfully."
        )

        return normalized_df
```

```python
# ============================================================
# CATEGORICAL ENCODING
# ============================================================

class DataInspector(DataInspector):

    def extract_normalized_categorical_data(
        self,
        method="onehot"
    ):

        """
        Encode categorical variables using:
        - OneHot Encoding
        - Ordinal Encoding
        - Uniform Encoding
        """

        categorical_data = self.df.select_dtypes(
            exclude=np.number
        )

        if categorical_data.empty:

            print("No categorical columns found.")

            return pd.DataFrame()

        if method.lower() == "onehot":

            encoder = OneHotEncoder(
                sparse_output=False,
                handle_unknown="ignore"
            )

            encoded = encoder.fit_transform(
                categorical_data
            )

            encoded_df = pd.DataFrame(
                encoded,
                columns=encoder.get_feature_names_out(
                    categorical_data.columns
                )
            )

        elif method.lower() == "ordinal":

            encoder = OrdinalEncoder()

            encoded = encoder.fit_transform(
                categorical_data
            )

            encoded_df = pd.DataFrame(
                encoded,
                columns=categorical_data.columns
            )

        elif method.lower() == "uniform":

            encoded_df = pd.DataFrame()

            for column in categorical_data.columns:

                codes = (
                    categorical_data[column]
                    .astype("category")
                    .cat.codes
                )

                encoded_df[column] = (
                    (codes - codes.min())
                    /
                    (codes.max() - codes.min())
                )

        else:

            raise ValueError(
                "Choose onehot, ordinal or uniform"
            )

        print(
            f"{method.upper()} encoding completed successfully."
        )

        return encoded_df
```

```python
# ============================================================
# MERGE NORMALIZED DATA
# ============================================================

class DataInspector(DataInspector):

    def create_normalized_data_df(self):

        """
        Create a final normalized dataset
        by combining numerical and
        categorical features.
        """

        normalized_numeric = (
            self.extract_normalized_numeric_data(
                method="robust"
            )
        )

        encoded_categorical = (
            self.extract_normalized_categorical_data(
                method="onehot"
            )
        )

        final_df = pd.concat(
            [
                normalized_numeric,
                encoded_categorical
            ],
            axis=1
        )

        print(
            "Unified normalized dataset created."
        )

        return final_df
```

# Advanced Interactive Visualization

Visualization is a crucial part of Exploratory Data Analysis (EDA).

### Relationship Analysis

The engine automatically selects the most suitable visualization based on the variable types.

| Variable Types | Visualization |
|---------------|--------------|
| Numeric vs Numeric | Scatter Plot |

These visualizations help identify distributions, trends, relationships and potential anomalies within the dataset.

```python
# ============================================================
# NUMERICAL VISUALIZATION
# ============================================================

class DataInspector(DataInspector):

    def plot_numerical(self, columns):

        """
        Generate:
        - Violin Plot
        - Scatter Plot
        - Histogram

        for selected numerical columns.
        """

        for column in columns:

            fig = make_subplots(
                rows=1,
                cols=3,
                subplot_titles=[
                    "Violin Plot",
                    "Scatter Plot",
                    "Histogram"
                ]
            )

            fig.add_trace(
                go.Violin(
                    y=self.df[column],
                    name=column,
                    box_visible=True
                ),
                row=1,
                col=1
            )

            fig.add_trace(
                go.Scatter(
                    y=self.df[column],
                    mode="markers",
                    name=column
                ),
                row=1,
                col=2
            )

            fig.add_trace(
                go.Histogram(
                    x=self.df[column],
                    name=column
                ),
                row=1,
                col=3
            )

            fig.update_layout(
                title=f"Numerical Analysis : {column}",
                height=500,
                width=1200
            )

            fig.show()
```

```python
# ============================================================
# RELATIONSHIP ANALYSIS
# ============================================================

class DataInspector(DataInspector):

    def plot_relationship(
        self,
        column_1,
        column_2
    ):

        """
        Automatically select the
        appropriate visualization.

        Numeric vs Numeric
            -> Scatter Plot

        Categorical vs Numeric
            -> Box Plot

        Categorical vs Categorical
            -> Grouped Bar Chart
        """

        col1_numeric = pd.api.types.is_numeric_dtype(
            self.df[column_1]
        )

        col2_numeric = pd.api.types.is_numeric_dtype(
            self.df[column_2]
        )

        if col1_numeric and col2_numeric:

            fig = px.scatter(
                self.df,
                x=column_1,
                y=column_2,
                trendline="ols",
                title=f"{column_1} vs {column_2}"
            )

        elif col1_numeric != col2_numeric:

            if col1_numeric:

                fig = px.box(
                    self.df,
                    x=column_2,
                    y=column_1,
                    title=f"{column_1} vs {column_2}"
                )

            else:

                fig = px.box(
                    self.df,
                    x=column_1,
                    y=column_2,
                    title=f"{column_1} vs {column_2}"
                )

        else:

            fig = px.histogram(
                self.df,
                x=column_1,
                color=column_2,
                barmode="group",
                title=f"{column_1} vs {column_2}"
            )

        fig.show()
```

# Deep Statistical Insights

Understanding relationships between variables is essential for data exploration and feature engineering.

A unified association heatmap is generated to analyze relationships among all variables within the dataset.

The following statistical measures are used:

### Numerical vs Numerical
- Pearson Correlation

### Categorical vs Categorical
- Cramer's V

### Numerical vs Categorical
- Eta Squared

The generated heatmap provides a comprehensive overview of feature associations within the dataset.

```python
# ============================================================
# UNIFIED ASSOCIATION HEATMAP
# ============================================================

class DataInspector(DataInspector):

    def plot_all_associations_heatmap(self):

        """
        Generate a unified association heatmap.

        Numerical-Numerical:
            Pearson Correlation

        Categorical-Categorical:
            Cramer's V

        Numerical-Categorical:
            Eta Squared
        """

        columns = self.df.columns

        association_matrix = pd.DataFrame(
            index=columns,
            columns=columns
        )

        for col1 in columns:

            for col2 in columns:

                try:

                    col1_numeric = pd.api.types.is_numeric_dtype(
                        self.df[col1]
                    )

                    col2_numeric = pd.api.types.is_numeric_dtype(
                        self.df[col2]
                    )

                    # Numeric vs Numeric
                    if col1_numeric and col2_numeric:

                        association_matrix.loc[
                            col1,
                            col2
                        ] = self.df[
                            [col1, col2]
                        ].corr().iloc[0, 1]

                    # Categorical vs Categorical
                    elif (
                        not col1_numeric
                        and
                        not col2_numeric
                    ):

                        association_matrix.loc[
                            col1,
                            col2
                        ] = cramers_v(
                            self.df[col1],
                            self.df[col2]
                        )

                    # Numeric vs Categorical
                    else:

                        if col1_numeric:

                            association_matrix.loc[
                                col1,
                                col2
                            ] = eta_squared(
                                self.df[col2],
                                self.df[col1]
                            )

                        else:

                            association_matrix.loc[
                                col1,
                                col2
                            ] = eta_squared(
                                self.df[col1],
                                self.df[col2]
                            )

                except:

                    association_matrix.loc[
                        col1,
                        col2
                    ] = 0

        fig = px.imshow(
            association_matrix.astype(float),
            text_auto=True,
            aspect="auto",
            title="Unified Association Heatmap"
        )

        fig.show()
```
