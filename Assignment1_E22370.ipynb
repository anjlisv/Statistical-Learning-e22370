## Q1: Solution

Let the disk have radius 10. Define the event

$$
A = \{\text{point lies within 1 unit of the boundary}\}
$$

So the distance from the center satisfies

$$
9 \le r \le 10
$$


### Model 1. Uniform distribution with respect to area

If the point is chosen uniformly over the disk, then $(X,Y)$ is a continuous random variable with constant PDF.

From probability theory:

$$
P(A) = \iint_A f(x,y)\,dx\,dy
$$

Since the distribution is uniform over the disk,

$$
f(x,y) = \frac{1}{\text{Area of disk}} = \frac{1}{100\pi}
$$

Thus,

$$
P(A) = \frac{\text{Area of }A}{100\pi}
$$

Area of disk:

$$
100\pi
$$

Area of region $A$ (annulus between 9 and 10):

$$
100\pi - 81\pi = 19\pi
$$

Therefore,

$$
P(A) = \frac{19\pi}{100\pi} = \frac{19}{100}
$$


### Model 2. Uniform distribution of the radius

Here, $r \sim \text{Uniform}(0,10)$

$$
f(r) = \frac{1}{10}, \quad 0 \le r \le 10
$$

$$
P(A) = \int_9^{10} \frac{1}{10} dr = \frac{1}{10}
$$


###  Reason for the Difference

The two answers are different because the phrase  
“a point is chosen at random from a disk” does not define a unique probability model.


### Model 1: Uniform distribution over area

In this case, the point is chosen uniformly over the **area of the disk**.

So probability is calculated using a joint density:

$$
P(A) = \iint_A f(x,y)\,dx\,dy
$$

where the density is constant:

$$
f(x,y) = \frac{1}{100\pi}
$$

Here, probability depends on **area**.  
The outer region of the disk (near the boundary) covers more area than the inner region.

Therefore, points near the boundary are more likely.

This gives:

$$
P(A) = \frac{19}{100}
$$


### Model 2: Uniform distribution over radius

In this case, the radius is uniformly distributed:

$$
r \sim \text{Uniform}(0,10)
$$

So probability is calculated using:

$$
P(A) = \int f(r)\,dr
$$

where

$$
f(r) = \frac{1}{10}
$$

Here, probability depends only on the **length of the interval**.

The interval $[9,10]$ has length 1 out of total length 10.

Thus:

$$
P(A) = \frac{1}{10}
$$


The difference comes from using **two different probability models**:

- One based on area (2D)
- One based on radius (1D)

Even though the event $A$ is the same in both cases,  
the way randomness is defined is different.

### Conclusion

Different probability models ⇒ different probabilities.

# Q2: Solution


Consider a unit circle:

$$
x^2 + y^2 = 1
$$

The side length of the inscribed equilateral triangle is:

$$
\sqrt{3}
$$

From geometry, a chord is longer than $\sqrt{3}$ if and only if its perpendicular distance from the center satisfies:

$$
r < \frac{1}{2}
$$

So the required event is:

$$
A = \{\text{distance of chord from center} < 1/2\}
$$


## Model 1: Uniform endpoint (angle model)

### Probability model

Let $\theta$ be the central angle subtended by the chord.

From probability theory:

- $\theta$ is a continuous random variable  

$$
P(A) = \int f(\theta)\, d\theta
$$

Since the angle is uniformly distributed:

$$
f(\theta) = \frac{1}{\pi}, \quad 0 \le \theta \le \pi
$$


### Event condition

Chord length:

$$
L(\theta) = 2\sin(\theta/2)
$$

We want:

$$
L(\theta) > \sqrt{3}
$$

$$
2\sin(\theta/2) > \sqrt{3}
$$

$$
\sin(\theta/2) > \frac{\sqrt{3}}{2}
$$

$$
\theta > \frac{2\pi}{3}
$$

So,

$$
A_1 = \left(\frac{2\pi}{3}, \pi\right]
$$


### Probability

$$
P_1(A) = \int_{2\pi/3}^{\pi} \frac{1}{\pi} d\theta
= \frac{\pi - 2\pi/3}{\pi}
= \frac{1}{3}
$$


## Model 2: Uniform distance from center

### Probability model

Let $r$ be the perpendicular distance from the center.

From the basic theories:

$$
P(A) = \int f(r)\, dr
$$

Since $r \sim \text{Uniform}(0,1)$:

$$
f(r) = 1
$$


### Event condition

$$
A_2 = [0, 1/2)
$$


### Probability

$$
P_2(A) = \int_{0}^{1/2} 1\, dr = \frac{1}{2}
$$


## Model 3: Uniform midpoint in the disk

### Probability model

Let $(X,Y)$ be the midpoint of the chord.

From probability theory :

$$
P(A) = \iint_A f(x,y)\, dx\, dy
$$

Since the midpoint is uniformly distributed over the disk:

$$
f(x,y) = \frac{1}{\pi}
$$


### Event condition

Distance from center:

$$
r = \sqrt{x^2 + y^2}
$$

Event:

$$
A_3 = \{x^2 + y^2 < (1/2)^2\}
$$


### Probability

$$
P_3(A) = \frac{\text{Area of smaller disk}}{\text{Area of unit disk}}
= \frac{\pi (1/2)^2}{\pi}
= \frac{1}{4}
$$


## Final Answers

$$
P_1(A) = \frac{1}{3}, \quad
P_2(A) = \frac{1}{2}, \quad
P_3(A) = \frac{1}{4}
$$


The probability of an event depends on the underlying probability distribution.

Here:

- Model 1 → uniform distribution over angle  
- Model 2 → uniform distribution over distance  
- Model 3 → uniform distribution over area  

Each model defines a different probability space, so the answers are different.

# Q8: Solution

A family has two children. Each child is equally likely to be a boy (B) or a girl (G), and the two outcomes are independent.


## Step 1: Define the sample space

From probability theory:

For equally likely outcomes,

$$
P(E) = \frac{\text{number of favourable outcomes}}{\text{total number of outcomes}}
$$

The sample space is:

$$
\Omega = \{BB, BG, GB, GG\}
$$

Each outcome has probability:

$$
P = \frac{1}{4}
$$


## Step 2: Define the given condition

We are given:

“At least one child is a girl”

This is a **conditional probability** problem.

Therefore ;

$$
P(A \mid B) = \frac{P(A \cap B)}{P(B)}
$$

Let:

- $A$: both children are girls → $\{GG\}$  
- $B$: at least one child is a girl  

Event $B$ excludes only $BB$, so:

$$
B = \{BG, GB, GG\}
$$


## Step 3: Compute the conditional probability

Using:

$$
P(A \mid B) = \frac{P(A \cap B)}{P(B)}
$$

- $A \cap B = \{GG\}$  
- $P(GG) = \frac{1}{4}$  

- $P(B) = \frac{3}{4}$  

Thus,

$$
P(A \mid B) = \frac{1/4}{3/4} = \frac{1}{3}
$$


## Final Answer

$$
P(\text{both girls} \mid \text{at least one girl}) = \frac{1}{3}
$$


From probability theory:

- This uses **conditional probability**  
- The result depends on the **given information (conditioning event)**  

If instead we were told: “the older child is a girl”,
then the sample space becomes:

$$
\{GG, GB\}
$$

So,

$$
P(\text{both girls}) = \frac{1}{2}
$$


## Conclusion

This problem shows that:

Probability depends on how the information is given  

Even though the situation is similar, different conditioning leads to different probabilities.

# Q9: Solution



1. From probability theory:

A probability space is defined as $(\Omega, \mathfrak{F}, \mathfrak{P})$

Given:

$$
\Omega = \{G_1, G_2, D_1, D_2\}, \quad \mathfrak{F} = \mathfrak{P}(\Omega)
$$

Probabilities:

$$
P(G_1)=0.50,\quad P(G_2)=0.20,\quad P(D_1)=0.10,\quad P(D_2)=0.20
$$

Since,

$$
0.50 + 0.20 + 0.10 + 0.20 = 1
$$

this is a valid probability measure.

---

2. Random variable $X$:

$$
X(G_1)=0,\quad X(G_2)=0,\quad X(D_1)=1,\quad X(D_2)=1
$$

From theory:

$\sigma(X)$ is generated by inverse images of values of $X$

$$
X^{-1}(0) = \{G_1, G_2\}, \quad X^{-1}(1) = \{D_1, D_2\}
$$

Thus,

$$
\sigma(X) = \{\emptyset, \Omega, \{G_1, G_2\}, \{D_1, D_2\}\}
$$

---

3. Random variable $Y$:

$$
Y(G_1)=1,\quad Y(G_2)=2,\quad Y(D_1)=3,\quad Y(D_2)=4
$$

Since each outcome has a unique value:

$$
\sigma(Y) = \mathfrak{P}(\Omega)
$$

---

4. Every set in $\sigma(X)$ is also a subset of $\Omega$, and since $\sigma(Y) = \mathcal{P}(\Omega)$, it contains all possible subsets of $\Omega$.

Therefore,

$$
\sigma(X) \subseteq \sigma(Y)
$$


This means that $Y$ carries more information than $X$.

- $X$ only tells us whether the item is **good or defective**
- $Y$ tells us both the **quality** and **which machine** produced it

Since $\sigma(Y)$ includes more detailed events,  
$Y$ is more informative than $X$.

---

5. The marginal distribution of $X$ is
$$
P(X=0) = P(G_1) + P(G_2) = 0.50 + 0.20 = 0.70
$$

$$
P(X=1) = P(D_1) + P(D_2) = 0.10 + 0.20 = 0.30
$$

---

6. Marginal distribution of $Y$ is

$$
P(Y=1)=0.50,\quad P(Y=2)=0.20,\quad P(Y=3)=0.10,\quad P(Y=4)=0.20
$$

---

7. From theory:

Joint probability is:

$$
P(X=x, Y=y) = P(\text{corresponding outcome})
$$

Thus,

$$
P(0,1)=0.50,\quad P(0,2)=0.20,\quad P(1,3)=0.10,\quad P(1,4)=0.20
$$

All other joint probabilities are zero.

---

8. From theory:

$$
P(A \mid B) = \frac{P(A \cap B)}{P(B)}
$$

$$
P(X=1 \mid Y=3) = 1,\quad P(X=1 \mid Y=4) = 1
$$

$$
P(Y=3 \mid X=1) = \frac{0.10}{0.30} = \frac{1}{3}
$$

$$
P(Y=4 \mid X=1) = \frac{0.20}{0.30} = \frac{2}{3}
$$

---

9.Conditioning on $Y$ gives more detailed information than conditioning on $X$.

If we know $Y = 3$, then we know exactly which outcome occurred.  
In this case, the item is definitely defective, so:

$$
P(X = 1 \mid Y = 3) = 1
$$

However, if we only know that $X = 1$, we only know that the item is defective,  
but we do not know which machine produced it.

So there is still some uncertainty:

$$
P(Y = 3 \mid X = 1) = \frac{1}{3}, \quad
P(Y = 4 \mid X = 1) = \frac{2}{3}
$$

Knowing $Y$ determines the exact outcome,  
while knowing $X$ gives only partial information.

Therefore, $Y$ provides more information than $X$.

---

10. The $\sigma$-algebra $\sigma(X)$ contains only information about the **quality class** of the item:

- whether the item is **good**, or  
- whether the item is **defective**  

It does **not distinguish** between Machine 1 and Machine 2.


In contrast, the $\sigma$-algebra $\sigma(Y)$ contains **full information about the outcome**. It tells us:

- whether the item is good or defective, and  
- which machine (Machine 1 or Machine 2) produced it  


Thus, $\sigma(Y)$ is **strictly finer** than $\sigma(X)$.


$$
\sigma(X) \subset \sigma(Y)
$$

Therefore, the random variable $Y$ contains more information than $X$.

---

11. Shannon entropy:

$$
H(Z) = -\sum P(z)\ln P(z)
$$

$$
H(X) = -(0.70\ln0.70 + 0.30\ln0.30) \approx 0.6109
$$

$$
H(Y) = -(0.50\ln0.50 + 0.20\ln0.20 + 0.10\ln0.10 + 0.20\ln0.20) \approx 1.2206
$$

$$
H(X,Y) = H(Y) \approx 1.2206
$$

$$
H(X \mid Y) = 0, \quad H(Y \mid X) = H(Y) - H(X) \approx 0.6097
$$


- $Y$ completely determines $X$  
- $X$ does not determine $Y$  

Therefore,

$$
\text{$Y$ contains more information than $X$}
$$

# Q10: Solution


1.From probability theory:

A continuous probability space is $(\Omega, \mathcal{F}, P)$, where probabilities are defined using a probability density function (PDF):

$$
P(A) = \iint_A f_{T,P}(t,p)\, dt\, dp
$$

Given:

$$
\Omega = \mathbb{R}^2, \quad \mathcal{F} = \mathcal{B}(\mathbb{R}^2)
$$

A point $\omega \in \Omega$ is:

$$
\omega = (t,p)
$$

The joint density is:

$$
f_{T,P}(t,p) = \frac{1}{4\pi} \exp\left(-\frac{t^2}{2} - \frac{p^2}{8}\right)
$$

Since $T \sim N(0,1)$ and $P \sim N(0,4)$ are independent, this defines a valid probability model.

---

2. The random variable $X$ is defined by

$$
X(t,p) = t
$$

Since

$$
\sigma(X) = \{X^{-1}(B) : B \in \mathcal{B}(\mathbb{R})\}
$$

$$
X^{-1}(B) = \{(t,p) \in \mathbb{R}^2 : t \in B\} = B \times \mathbb{R}
$$

Thus,

$$
\sigma(X) = \{B \times \mathbb{R} : B \in \mathcal{B}(\mathbb{R})\}
$$
These are vertical strips (depend only on $t$).

---

3. The random variable $Y$ is defined by

$$
Y(t,p) = (t,p)
$$

Since $Y$ is the identity map:

$$
\sigma(Y) = \mathcal{B}(\mathbb{R}^2)
$$

---

4. Since every set of the form $B\times\mathbb R$ is a Borel subset of $\mathbb R^2$, we have

$$
\sigma(X) \subset \sigma(Y)
$$

This is strict because sets like:

$$
[-1,1] \times [-2,2]
$$

belong to $\sigma(Y)$ but not to $\sigma(X)$.

Hence, $Y$ contains more information than $X$.

Thus,

$$
X \sim N(0,1)
$$

This is a bivariate normal distribution with independent components.

$$

and knowing $Y$ determines $X$, but not vice versa.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Circle

fig, axes = plt.subplots(1, 3, figsize=(15, 4.8))

xlim = (-4, 4)
ylim = (-8, 8)

# -------------------------
# (1) Event in sigma(X)
# -------------------------
ax = axes[0]
ax.set_title(r"Event in $\sigma(X)$: $B\times\mathbb{R}$")
ax.set_xlim(xlim)
ax.set_ylim(ylim)
ax.set_xlabel(r"$t$")
ax.set_ylabel(r"$p$")

# Example B = [-1,1]
strip = Rectangle((-1, ylim[0]), 2, ylim[1] - ylim[0],
                  facecolor="skyblue", edgecolor="navy", alpha=0.55)
ax.add_patch(strip)

ax.text(0, 0, r"$[-1,1]\times \mathbb{R}$",
        ha="center", va="center", fontsize=12,
        bbox=dict(facecolor="white", edgecolor="none", alpha=0.8))

ax.axhline(0, color="black", lw=0.8)
ax.axvline(0, color="black", lw=0.8)
ax.grid(alpha=0.25)

# -------------------------
# (2) Event in sigma(Y)
# -------------------------
ax = axes[1]
ax.set_title(r"Event in $\sigma(Y)$: general Borel subset")
ax.set_xlim(xlim)
ax.set_ylim(ylim)
ax.set_xlabel(r"$t$")
ax.set_ylabel(r"$p$")

# Example rectangle in sigma(Y)
rect = Rectangle((-1.5, -3), 3, 6,
                 facecolor="lightgreen", edgecolor="darkgreen", alpha=0.55)
ax.add_patch(rect)

# Also add a disk to emphasize generality
disk = Circle((2.2, 3.5), 1.2,
              facecolor="salmon", edgecolor="darkred", alpha=0.45)
ax.add_patch(disk)

ax.text(0, 0, r"$[-1.5,1.5]\times[-3,3]$",
        ha="center", va="center", fontsize=11,
        bbox=dict(facecolor="white", edgecolor="none", alpha=0.8))
ax.text(2.2, 3.5, r"disk",
        ha="center", va="center", fontsize=10,
        bbox=dict(facecolor="white", edgecolor="none", alpha=0.8))

ax.axhline(0, color="black", lw=0.8)
ax.axvline(0, color="black", lw=0.8)
ax.grid(alpha=0.25)

# -------------------------
# (3) Comparison
# -------------------------
ax = axes[2]
ax.set_title(r"Comparison: $\sigma(X)\subseteq\sigma(Y)$")
ax.set_xlim(xlim)
ax.set_ylim(ylim)
ax.set_xlabel(r"$t$")
ax.set_ylabel(r"$p$")

# Vertical strip for sigma(X)
strip2 = Rectangle((-1, ylim[0]), 2, ylim[1] - ylim[0],
                   facecolor="skyblue", edgecolor="navy", alpha=0.35,
                   label=r"$\sigma(X)$ event")
ax.add_patch(strip2)

# Rectangle for sigma(Y)
rect2 = Rectangle((-1.5, -3), 3, 6,
                  facecolor="lightgreen", edgecolor="darkgreen", alpha=0.5,
                  label=r"$\sigma(Y)$ event")
ax.add_patch(rect2)

ax.text(0, 6.5, r"$\sigma(X)$ sees only vertical strips",
        ha="center", fontsize=11,
        bbox=dict(facecolor="white", edgecolor="none", alpha=0.8))
ax.text(0, -6.5, r"$\sigma(Y)$ sees full 2D regions",
        ha="center", fontsize=11,
        bbox=dict(facecolor="white", edgecolor="none", alpha=0.8))

ax.axhline(0, color="black", lw=0.8)
ax.axvline(0, color="black", lw=0.8)
ax.legend(loc="upper right")
ax.grid(alpha=0.25)

plt.tight_layout()
plt.show()
```
