## Q1: Solution

Let the disk have radius 10. Define the event

$$
A = \{\text{point lies within 1 unit of the boundary}\}
$$

So the distance from the center satisfies

$$
9 \le r \le 10
$$

---

### 1. Uniform distribution with respect to area

If the point is chosen uniformly over the disk, then $(X,Y)$ is a continuous random variable with constant PDF.

From probability theory:

$$
P(A) = \iint_A f(x,y)\,dx\,dy
$$

Since the distribution is uniform over the disk,

$$
f(x,y) = \frac{1}{\text{Area of disk}} = \frac{1}{100\pi}
$$

Thus,

$$
P(A) = \frac{\text{Area of }A}{100\pi}
$$

Area of disk:

$$
100\pi
$$

Area of region $A$ (annulus between 9 and 10):

$$
100\pi - 81\pi = 19\pi
$$

Therefore,

$$
P(A) = \frac{19\pi}{100\pi} = \frac{19}{100}
$$

---

### 2. Uniform distribution of the radius

Here, $r \sim \text{Uniform}(0,10)$

$$
f(r) = \frac{1}{10}, \quad 0 \le r \le 10
$$

$$
P(A) = \int_9^{10} \frac{1}{10} dr = \frac{1}{10}
$$

---

### 3. Reason for the difference

- Uniform over area:
$$
P(A) = \iint_A f(x,y)\,dx\,dy
$$

- Uniform over radius:
$$
P(A) = \int f(r)\,dr
$$

Different probability models → different answers.

---

### Final Answer

$$
P(A) = \frac{19}{100} \quad (\text{uniform over area})
$$

$$
P(A) = \frac{1}{10} \quad (\text{uniform over radius})
$$

# Q2: Solution


Consider a unit circle:

$$
x^2 + y^2 = 1
$$

The side length of the inscribed equilateral triangle is:

$$
\sqrt{3}
$$

From geometry, a chord is longer than $\sqrt{3}$ if and only if its perpendicular distance from the center satisfies:

$$
r < \frac{1}{2}
$$

So the required event is:

$$
A = \{\text{distance of chord from center} < 1/2\}
$$

---

## Model 1: Uniform endpoint (angle model)

### Probability model

Let $\theta$ be the central angle subtended by the chord.

From probability theory:

- $\theta$ is a continuous random variable  

$$
P(A) = \int f(\theta)\, d\theta
$$

Since the angle is uniformly distributed:

$$
f(\theta) = \frac{1}{\pi}, \quad 0 \le \theta \le \pi
$$

---

### Event condition

Chord length:

$$
L(\theta) = 2\sin(\theta/2)
$$

We want:

$$
L(\theta) > \sqrt{3}
$$

$$
2\sin(\theta/2) > \sqrt{3}
$$

$$
\sin(\theta/2) > \frac{\sqrt{3}}{2}
$$

$$
\theta > \frac{2\pi}{3}
$$

So,

$$
A_1 = \left(\frac{2\pi}{3}, \pi\right]
$$

---

### Probability

$$
P_1(A) = \int_{2\pi/3}^{\pi} \frac{1}{\pi} d\theta
= \frac{\pi - 2\pi/3}{\pi}
= \frac{1}{3}
$$

---

## Model 2: Uniform distance from center

### Probability model

Let $r$ be the perpendicular distance from the center.

From the basic theories:

$$
P(A) = \int f(r)\, dr
$$

Since $r \sim \text{Uniform}(0,1)$:

$$
f(r) = 1
$$

---

### Event condition

$$
A_2 = [0, 1/2)
$$

---

### Probability

$$
P_2(A) = \int_{0}^{1/2} 1\, dr = \frac{1}{2}
$$

---

## Model 3: Uniform midpoint in the disk

### Probability model

Let $(X,Y)$ be the midpoint of the chord.

From probability theory :

$$
P(A) = \iint_A f(x,y)\, dx\, dy
$$

Since the midpoint is uniformly distributed over the disk:

$$
f(x,y) = \frac{1}{\pi}
$$

---

### Event condition

Distance from center:

$$
r = \sqrt{x^2 + y^2}
$$

Event:

$$
A_3 = \{x^2 + y^2 < (1/2)^2\}
$$

---

### Probability

$$
P_3(A) = \frac{\text{Area of smaller disk}}{\text{Area of unit disk}}
= \frac{\pi (1/2)^2}{\pi}
= \frac{1}{4}
$$

---

## Final Answers

$$
P_1(A) = \frac{1}{3}, \quad
P_2(A) = \frac{1}{2}, \quad
P_3(A) = \frac{1}{4}
$$

---

## Conclusion

The probability of an event depends on the underlying probability distribution.

Here:

- Model 1 → uniform distribution over angle  
- Model 2 → uniform distribution over distance  
- Model 3 → uniform distribution over area  

Each model defines a different probability space, so the answers are different.

# Q8: Probability of Being a Girl

## Solution

A family has two children. Each child is equally likely to be a boy (B) or a girl (G), and the two outcomes are independent.

---

## Step 1: Define the sample space

From probability theory:

For equally likely outcomes,

$$
P(E) = \frac{\text{number of favourable outcomes}}{\text{total number of outcomes}}
$$

The sample space is:

$$
\Omega = \{BB, BG, GB, GG\}
$$

Each outcome has probability:

$$
P = \frac{1}{4}
$$

---

## Step 2: Define the given condition

We are given:

“At least one child is a girl”

This is a **conditional probability** problem.

Therefore ;

$$
P(A \mid B) = \frac{P(A \cap B)}{P(B)}
$$

Let:

- $A$: both children are girls → $\{GG\}$  
- $B$: at least one child is a girl  

Event $B$ excludes only $BB$, so:

$$
B = \{BG, GB, GG\}
$$

---

## Step 3: Compute the conditional probability

Using:

$$
P(A \mid B) = \frac{P(A \cap B)}{P(B)}
$$

- $A \cap B = \{GG\}$  
- $P(GG) = \frac{1}{4}$  

- $P(B) = \frac{3}{4}$  

Thus,

$$
P(A \mid B) = \frac{1/4}{3/4} = \frac{1}{3}
$$

---

## Final Answer

$$
P(\text{both girls} \mid \text{at least one girl}) = \frac{1}{3}
$$

---

## Explanation
From probability theory:

- This uses **conditional probability**  
- The result depends on the **given information (conditioning event)**  

If instead we were told: “the older child is a girl”,
then the sample space becomes:

$$
\{GG, GB\}
$$

So,

$$
P(\text{both girls}) = \frac{1}{2}
$$

---

## Conclusion

This problem shows that:

👉 Probability depends on how the information is given  

Even though the situation is similar, different conditioning leads to different probabilities.

# Q9: Solution

---

## 1. Probability Space

From probability theory:

A probability space is defined as $(\Omega, \mathcal{F}, P)$

Given:

$$
\Omega = \{G_1, G_2, D_1, D_2\}, \quad \mathcal{F} = \mathcal{P}(\Omega)
$$

Probabilities:

$$
P(G_1)=0.50,\quad P(G_2)=0.20,\quad P(D_1)=0.10,\quad P(D_2)=0.20
$$

Since,

$$
0.50 + 0.20 + 0.10 + 0.20 = 1
$$

this is a valid probability measure.

---

## 2. $\sigma$-algebra generated by $X$

Random variable $X$:

$$
X(G_1)=0,\quad X(G_2)=0,\quad X(D_1)=1,\quad X(D_2)=1
$$

From theory:

$\sigma(X)$ is generated by inverse images of values of $X$

$$
X^{-1}(0) = \{G_1, G_2\}, \quad X^{-1}(1) = \{D_1, D_2\}
$$

Thus,

$$
\sigma(X) = \{\emptyset, \Omega, \{G_1, G_2\}, \{D_1, D_2\}\}
$$

---

## 3. $\sigma$-algebra generated by $Y$

Random variable $Y$:

$$
Y(G_1)=1,\quad Y(G_2)=2,\quad Y(D_1)=3,\quad Y(D_2)=4
$$

Since each outcome has a unique value:

$$
\sigma(Y) = \mathcal{P}(\Omega)
$$

---

## 4. Relation between $\sigma$-algebras

$$
\sigma(X) \subseteq \sigma(Y)
$$

From theory: a finer $\sigma$-algebra contains more information.

Thus, $Y$ contains more information than $X$.

---

## 5. Marginal distribution of $X$

$$
P(X=0) = P(G_1) + P(G_2) = 0.50 + 0.20 = 0.70
$$

$$
P(X=1) = P(D_1) + P(D_2) = 0.10 + 0.20 = 0.30
$$

---

## 6. Marginal distribution of $Y$

$$
P(Y=1)=0.50,\quad P(Y=2)=0.20,\quad P(Y=3)=0.10,\quad P(Y=4)=0.20
$$

---

## 7. Joint probabilities

From theory:

Joint probability is:

$$
P(X=x, Y=y) = P(\text{corresponding outcome})
$$

Thus,

$$
P(0,1)=0.50,\quad P(0,2)=0.20,\quad P(1,3)=0.10,\quad P(1,4)=0.20
$$

All other joint probabilities are zero.

---

## 8. Conditional probabilities

From theory:

$$
P(A \mid B) = \frac{P(A \cap B)}{P(B)}
$$

$$
P(X=1 \mid Y=3) = 1,\quad P(X=1 \mid Y=4) = 1
$$

$$
P(Y=3 \mid X=1) = \frac{0.10}{0.30} = \frac{1}{3}
$$

$$
P(Y=4 \mid X=1) = \frac{0.20}{0.30} = \frac{2}{3}
$$

---

## 9. Interpretation

- Conditioning on $Y$ gives exact information  
- Conditioning on $X$ leaves uncertainty  

Hence, $Y$ is more informative.

---

## 10. Information available in $\sigma(X)$ and $\sigma(Y)$

The $\sigma$-algebra $\sigma(X)$ contains only information about the **quality class** of the item:

- whether the item is **good**, or  
- whether the item is **defective**  

It does **not distinguish** between Machine 1 and Machine 2.


In contrast, the $\sigma$-algebra $\sigma(Y)$ contains **full information about the outcome**. It tells us:

- whether the item is good or defective, and  
- which machine (Machine 1 or Machine 2) produced it  


Thus, $\sigma(Y)$ is **strictly finer** than $\sigma(X)$.



### Conclusion

$$
\sigma(X) \subset \sigma(Y)
$$

Therefore, the random variable $Y$ contains more information than $X$.

---

## 11. Entropy (Information Measure)

Shannon entropy:

$$
H(Z) = -\sum P(z)\ln P(z)
$$

$$
H(X) = -(0.70\ln0.70 + 0.30\ln0.30) \approx 0.6109
$$

$$
H(Y) = -(0.50\ln0.50 + 0.20\ln0.20 + 0.10\ln0.10 + 0.20\ln0.20) \approx 1.2206
$$

$$
H(X,Y) = H(Y) \approx 1.2206
$$

$$
H(X \mid Y) = 0, \quad H(Y \mid X) = H(Y) - H(X) \approx 0.6097
$$



## Conclusion

- $Y$ completely determines $X$  
- $X$ does not determine $Y$  

Therefore,

$$
\text{$Y$ contains more information than $X$}
$$

```python
import numpy as np
import matplotlib.pyplot as plt

# Sample space
states = ["G1", "G2", "D1", "D2"]
prob = np.array([0.50, 0.20, 0.10, 0.20])

# Random variables
X = [0, 0, 1, 1]
Y = [1, 2, 3, 4]

# Marginals
P_X = [prob[0] + prob[1], prob[2] + prob[3]]
P_Y = prob

# Joint matrix
joint = np.array([
    [0.50, 0.20, 0.00, 0.00],
    [0.00, 0.00, 0.10, 0.20]
])

# Plot joint heatmap
plt.imshow(joint, cmap='viridis')
plt.colorbar(label='Probability')
plt.xticks([0,1,2,3], [1,2,3,4])
plt.yticks([0,1], [0,1])
plt.xlabel("Y")
plt.ylabel("X")
plt.title("Joint Distribution of (X,Y)")

for i in range(2):
    for j in range(4):
        plt.text(j, i, f"{joint[i,j]:.2f}", ha='center', va='center')

plt.show()

# Bar plot for marginals
plt.bar(["X=0","X=1"], P_X)
plt.title("Marginal Distribution of X")
plt.show()

plt.bar(["1","2","3","4"], P_Y)
plt.title("Marginal Distribution of Y")
plt.show()
```

# Q10: Solution

---

## 1. Probability space

From probability theory:

👉 A continuous probability space is $(\Omega, \mathcal{F}, P)$, where probabilities are defined using a probability density function (PDF):

$$
P(A) = \iint_A f_{T,P}(t,p)\, dt\, dp
$$

Given:

$$
\Omega = \mathbb{R}^2, \quad \mathcal{F} = \mathcal{B}(\mathbb{R}^2)
$$

A point $\omega \in \Omega$ is:

$$
\omega = (t,p)
$$

The joint density is:

$$
f_{T,P}(t,p) = \frac{1}{4\pi} \exp\left(-\frac{t^2}{2} - \frac{p^2}{8}\right)
$$

Since $T \sim N(0,1)$ and $P \sim N(0,4)$ are independent, this defines a valid probability model.

---

## 2. $\sigma$-algebra generated by $X$

$$
X(t,p) = t
$$

From the lecture note:

$$
\sigma(X) = \{X^{-1}(B) : B \in \mathcal{B}(\mathbb{R})\}
$$

$$
X^{-1}(B) = \{(t,p) \in \mathbb{R}^2 : t \in B\} = B \times \mathbb{R}
$$

Thus,

$$
\sigma(X) = \{B \times \mathbb{R} : B \in \mathcal{B}(\mathbb{R})\}
$$

👉 These are vertical strips (depend only on $t$).

---

## 3. $\sigma$-algebra generated by $Y$

$$
Y(t,p) = (t,p)
$$

Since $Y$ is the identity map:

$$
\sigma(Y) = \mathcal{B}(\mathbb{R}^2)
$$

---

## 4. Relation between $\sigma$-algebras

$$
\sigma(X) \subset \sigma(Y)
$$

This is strict because sets like:

$$
[-1,1] \times [-2,2]
$$

belong to $\sigma(Y)$ but not to $\sigma(X)$.

👉 Hence, $Y$ contains more information than $X$.

---

## 5. Marginal density of $X$

From probability theory:

$$
f_X(x) = \int_{-\infty}^{\infty} f_{T,P}(x,p)\, dp
$$

$$
= \int_{-\infty}^{\infty} \frac{1}{4\pi} \exp\left(-\frac{x^2}{2} - \frac{p^2}{8}\right) dp
$$

Evaluating:

$$
f_X(x) = \frac{1}{\sqrt{2\pi}} e^{-x^2/2}
$$

Thus,

$$
X \sim N(0,1)
$$

---

## 6. Marginal density of $Y$

$$
f_Y(t,p) = f_{T,P}(t,p)
$$

👉 This is a bivariate normal distribution with independent components.

---

## 7. Required probabilities

$$
P(X \le 0) = \frac{1}{2}
$$

$$
P(X > 1) = 1 - \Phi(1) \approx 0.1587
$$

$$
P(Y \in (-\infty,0] \times \mathbb{R}) = \frac{1}{2}
$$

Since independence:

$$
P(Y \in [-1,1]\times[-2,2]) = (2\Phi(1)-1)^2 \approx 0.466
$$

---

## 8. Conditional distribution of $X \mid Y=(t,p)$

Since $X = t$:

$$
P(X = t \mid Y=(t,p)) = 1
$$

Thus,

$$
X \mid Y = \delta_t
$$

---

## 9. Conditional distribution of $Y \mid X=t$

$$
Y \mid X=t = (t,P), \quad P \sim N(0,4)
$$

$$
f_{Y|X}(u,v|t) = \delta_t(u)\cdot \frac{1}{\sqrt{8\pi}} e^{-v^2/8}
$$

---

## 10. Interpretation (Information)

- Conditioning on $X$: only temperature is known  
- Pressure remains random  

- Conditioning on $Y$: both temperature and pressure are known  

👉 Therefore, $Y$ provides more information.

---

## 11. Measurability relation

Define:

$$
\pi_1(t,p) = t
$$

Then:

$$
X = \pi_1 \circ Y
$$

Thus,

$$
\sigma(X) \subset \sigma(Y)
$$

---

## 12. Visualization interpretation

- $\sigma(X)$ → vertical strips $B \times \mathbb{R}$  
- $\sigma(Y)$ → all subsets of $\mathbb{R}^2$  

👉 $X$ captures one dimension, $Y$ captures full 2D information.

---

## 13. Differential entropy of $X$

From lecture note:

$$
h(X) = -\int f_X(x)\ln f_X(x)\,dx
$$

For normal distribution:

$$
h(X) = \frac{1}{2}\ln(2\pi e)
$$

$$
h(X) \approx 1.4189
$$

---
## 14. Differential Entropy of $Y$

Since $Y = (T,P)$ has independent normal components with variances $1$ and $4$, from the lecture note:

👉 Entropy of independent variables is additive:

$$
h(Y) = h(T) + h(P)
$$

Now,

$$
h(T) = \frac{1}{2}\ln(2\pi e)
$$

and

$$
h(P) = \frac{1}{2}\ln(2\pi e \cdot 4)
$$

Therefore,

$$
h(Y) = \frac{1}{2}\ln(2\pi e) + \frac{1}{2}\ln(8\pi e)
$$

Equivalently,

$$
h(Y) = \ln(2\pi e) + \ln 2
$$

Numerically,

$$
h(Y) \approx 3.5310
$$

---

## 15. Conditional Differential Entropy $h(Y \mid X)$

Since $T$ and $P$ are independent and $Y=(T,P)$:

👉 Knowing $X=T$ removes uncertainty in $T$, but not in $P$

Thus,

$$
h(Y \mid X) = h(P)
$$

$$
= \frac{1}{2}\ln(2\pi e \cdot 4)
$$

$$
= \frac{1}{2}\ln(8\pi e)
$$

Numerically,

$$
h(Y \mid X) \approx 2.1121
$$

👉 Interpretation:

Once temperature is known, the only remaining uncertainty is in pressure.

---

## 16. Behavior of $h(X \mid Y)$

Since $X$ is a deterministic function of $Y$:

$$
X = t
$$

One might expect:

$$
h(X \mid Y) = 0
$$

However, from continuous entropy theory:

👉 This is NOT correct.

In the continuous case:

- The conditional distribution of $X \mid Y$ is **degenerate** (Dirac delta)  
- Differential entropy is **not finite**  

Thus,

$$
h(X \mid Y) \to -\infty
$$

---

### Key Difference (Lecture Note Insight)

- Discrete case:
$$
H(X \mid Y) = 0
$$

- Continuous case:
$$
h(X \mid Y) = -\infty
$$

👉 Because exact knowledge corresponds to a **singular distribution**

---

## Final Interpretation

This example shows:

- $\sigma(X) \subset \sigma(Y)$  
- $Y$ contains more information than $X$  
- $h(Y \mid X) > 0$ (uncertainty remains)  
- $h(X \mid Y)$ behaves differently from the discrete case  

---

## Conclusion

The variables $X$ and $Y$ are defined on the same probability space, but:

- $X$ is a **coarser observation**  
- $Y$ is a **finer observation**

Thus,

$$
\sigma(X) \subset \sigma(Y)
$$

and knowing $Y$ determines $X$, but not vice versa.
